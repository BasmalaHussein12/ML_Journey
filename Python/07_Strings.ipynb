```python
# String Slicing
Sometimes we want part of a string

Syntax:
string[start:stop]

Python includes the start index but excludes the stop index.
```

```python
word = "Python"

print(word[0:3])
```

```python
word = "Python"

print(word[2:6])
```

```python
word = "Python"

print(word[:4])
```

```python
word = "Python"

print(word[2:])
```

```python
# Common String Methods

upper()
Converts all letters to uppercase

lower()
Converts all letters to lowercase

title()
Capitalizes the first letter of each word

capitalize()
Capitalizes only the first letter of the entire string
```

```python
name = "Basmala"

print(name.upper())
```

```python
name = "Basmala"

print(name.lower())
```

```python
name = "Basmala is an machine learning engineer"

print(name.title())
```

```python
name = "pyhton is very easy!"

print(name.capitalize())
```

```python
# Removing Extra Spaces

Suppose the user enters:
"   Basmala   "

The spaces can cause problems
Use strip()

Remove only left spaces
Usw lstrip()
```

```python
name = "   Basmala   "

print(name.strip())
```

```python
name = "   Basmala   "

print(name.lstrip())
```

```python
# split()
The split() method breaks a string into a list

Syntax:
string.split(separator)
```

```python
text = "Python Java C++"

languages = text.split()

print(languages)

# Python splits the string by spaces by default.
# Notice that the result is a list 
```

```python
text = "Apple,Banana,Orange"

fruits = text.split(",")

print(fruits)
```

```python
Why is split() useful?

Imagine the user enters:
Python Java C++ JavaScript

With split(), you can turn it into:
["Python", "Java", "C++", "JavaScript"]
Now you can loop through it easily.
```

```python
# join()
join() does the opposite of split()
It joins a list into a single string
```

```python
words = ["Machine", "Learning", "is", "fun"]

sentence = " ".join(words)

print(sentence)
```

```python
The " " means:
Put one space between each word.
```

```python
letters = ["P", "Y", "T", "H", "O", "N"]

print("-".join(letters))
```

```python
# f-Strings
I've already used them, but now let's understand them better
Instead of:
```

```python
name = "Basmala"

age = 19

print("My name is " + name + " and I am " + str(age) + " years old.")
```

```python
Use an f-string:
```

```python
name = "Basmala"
age = 19

print(f"My name is {name} and I am {age} years old.")
```

```python
# Expressions Inside f-Strings
You can even do calculations
```

```python
x = 5
y = 8

print(f"The sum is {x + y}")
```
