<!-- try and except -->
Instead of letting the program crash, we can catch the error.

```python
try: 
    numbers = int(input("Enter a number: "))
    print(number)

except: 
    print("Invalid input!")    
```

<!-- How Does It Work? -->

Think of try as saying:
"Python, try to run this code."

If everything is okay...
Continue normally.

If an error happens...
Jump to the except block.

<!-- Catching Specific Errors -->
Using only except: works, but it's better to catch specific errors.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Please enter numbers only.")
```

```python
x = 10
y = 0

print(x / y)
```

```python
try:
    print(x / y)
except ZeroDivisionError:
    print("You cannot divide by zero.")
```

<!-- So, I noticed that: -->
try:
    # Code that may cause an error
except:
    # Runs only if an error happens

But what if I want to run some code only if no error occurs?
That's what else is for.    

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("Great! You entered:", number)
```

try → Try this code.
except → Run if an error happens.
else → Run only if no error happens.

<!-- finally -->
Sometimes you have code that must run no matter what.
That's what finally is for.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
finally:
    print("Program finished.")
```

<!-- Why is finally useful? -->
Even if an error happens while reading, the file is still closed.
(Although nowadays, we usually use with open(), which does this automatically.)

```python
file = open("notes.txt", "r")

try:
    content = file.read()
finally:
    file.close()
```

<!-- Multiple except Blocks -->
A program can have more than one possible error.

```python
try:
    x = int(input("Enter the first number: "))
    y = int(input("Enter the second number: "))

    print(x / y)

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError:
    print("You cannot divide by zero.")
```

<!-- raise -->
Sometimes you want to create an error.
This is called raising an exception.

```python
age = int(input("Enter your age: "))

if age < 0:
    raise ValueError("Age cannot be negative.")
```

<!-- Summary -->
- try
Code that might cause an error

- except
Handles the error

- else
Runs if no error occurs

- finally
Runs whether an error occurs or not

- raise
Manually create an exception
