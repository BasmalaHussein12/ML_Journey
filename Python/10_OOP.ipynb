<!-- What is Encapsulation? -->
Encapsulation means:
Keeping data and the methods that work on it together, and protecting the data from being changed incorrectly.

<!-- Protecting Data -->
In Python, we can indicate that an attribute is internal by putting an underscore before its name.

```python
class bank_account:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balanse = balance
```

<!-- _balance -->
The single underscore is a convention that says:
"This attribute is for internal use. Please don't modify it directly."
It's not true security, but it's an important Python convention.

```python
class bank_account:

    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    acc1 = bank_account("Basmala", 10000)
    acc1.deposit(500)

    def withdraw(self, amount):
        if amount <= self._balance:
           self._balance -= amount
        else:
           print("Insufficient balance.")

    def show_balance(self):
               print(f"Current balance: {self._balance}")


    acc1.show_balance()         
```

<!-- Inheritance -->
First, let's understand the problem.
Imagine you have this class:

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name}.")
     <------------------------------------->

Now you want to create a Student.
Students also have:
Name 
Age 

But they also have:
GPA 

Should we rewrite everything?
class Student:

    def __init__(self, name, age, gpa):
        self.name = name
        self.age = age
        self.gpa = gpa

We're repeating code!
The Solution: Inheritance
A Student is a Person.
So instead of rewriting the code, the Student class can inherit from the Person class.

```python
# creating the parent class 

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi! I'm {self.name}.")

# creating the child class

class Student(Person):
    def __init__(self, name, age, gpa):
      super().__init__(name, age)
      self.gpa = gpa

student7 = Student("Basmala", 19, 3.34)
student7.introduce()  
print(student7.gpa)  
```

<!-- What is super()? -->
super() means:
"Go to my parent class."

So this line:
super().__init__(name, age)
calls the Person constructor to initialize name and age.

<!-- Why Use Inheritance? -->

Without inheritance:
You write the same code many times.

With inheritance:
Write it once.
Reuse it everywhere.
This makes your programs cleaner and easier to maintain.

<!-- What does "Polymorphism" mean? -->

The word comes from:
Poly = Many
Morph = Forms

So it basically means:
One interface, many different behaviors.

```python
class Dog:

    def speak(self):
        print("Woof!")

class Cat:

    def speak(self):
        print("Meow!")  

dog1 = Dog()
cat1 = Cat()  

dog1.speak()
cat1.speak()
```

- Both classes have a method called:
speak()

But each one behaves differently.
This is polymorphism.

```python
class Bird:

    def move(self):
        print("Flying")

class Fish:

    def move(self):
        print("Swimming")

bird = Bird()
fish = Fish()

bird.move()
fish.move()       
```

<!-- What is Abstraction? -->
Abstraction means:
Hide the complicated implementation and show only what the user needs to use.
The user doesn't need to know how something works internally—only how to use it.

<!-- Example in Python -->

Suppose we have a coffee machine.
- Without abstraction:
coffee_machine.heat_water()
coffee_machine.grind_beans()
coffee_machine.mix_everything()
coffee_machine.pour_into_cup()

The user has to know every step.

- With abstraction:
coffee_machine.make_coffee()
One simple method.

All the complicated work happens inside.
The user doesn't need to know the implementation.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def speak(self):
        pass
```

- ABC
It means:
This is an abstract class.
An abstract class is not meant to create objects directly.

- @abstractmethod
This means:
Every child class must implement this method.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):

    def speak(self):
        print("Woof!")

class Cat(Animal):

    def speak(self):
        print("Meow!")

dog = Dog()
dog.speak()
```

<!-- Encapsulation vs Abstraction -->

**Encapsulation
- Protects data
- Controls access to attributes
- Example: Private balance in a bank account

**Abstraction
- Hides complexity
- Shows only essential features
- Example: make_coffee() hides all the steps
