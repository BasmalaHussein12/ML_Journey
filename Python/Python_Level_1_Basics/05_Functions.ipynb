```python
# What is a Function?

Before functions, imagine you need to print:
<<Welcome to our Python course!>>
10 times

You could write:

print("Welcome to our Python course!")
print("Welcome to our Python course!")
print("Welcome to our Python course!")

...and repeat it

But that's repetitive

A function lets you write the code once and use it whenever you need it

Think of a function like a recipe:

You write the recipe once
You can use it many times
```

```python
# Creating my First Function

In Python, we create a function using the keyword:
def

-The syntax is:
def function_name():
    # code
```

```python
def greet():
    print("Hello!")
```

```python
Let's understand it:

def → tells Python you're defining a function.
greet → the function's name.
() → parentheses (they're required).
: → starts the function body.
The indented code is what the function does.

    <look here>
If you run only this:

def greet():
    print("Hello!")

Nothing is printed.

Why?

Because you've defined the function, but you haven't (called) it yet

Python doesn't execute this function
It only stores it in memory and remembers:
"Okay, I have a function called greet(). I'll run it whenever someone calls it."
```

```python
# Calling a Function

- To use the function:

greet()

- Output:

Hello!

Now the function runs.

Calling It Multiple Times
def greet():
    print("Hello!")

greet()
greet()
greet()

Output:

Hello!
Hello!
Hello!

One function, three calls.

Why Are Functions Useful?

Imagine you have this code in 20 places:

print("Welcome!")

Later, you decide to change it to:

Welcome to Python!

Without functions, you'd have to edit all 20 places.

With a function:

def welcome():
    print("Welcome!")

You only change it once, and every call uses the updated version.
```

```python
# Functions with Parameters
So far, our function always prints the same thing.

def greet():
    print("Hello!")

But what if we want to greet different people?

Instead of creating:

def greet_basmala():
    print("Hello, Basmala!")

def greet_ali():
    print("Hello, Ali!")

def greet_sara():
    print("Hello, Sara!")

We can create one function that works for everyone.
                                     
--------------------------------------------------------
                                     
<Parameters>
A parameter is a variable that belongs to the function.

def greet(name):
    print(f"Hello, {name}!")

Here:
name is called a parameter.

Calling the Function
greet("Basmala")

Output:

Hello, Basmala!

Call it again:

greet("Ali")
greet("Sara")

Output:

Hello, Ali!
Hello, Sara!

Then one function, different results.

<<How Does Python Know?>>

When you write:

greet("Basmala")

Python temporarily does something like this:

name = "Basmala"

Then executes:

print(f"Hello, {name}!")

So the output becomes:

Hello, Basmala!

After the function finishes, that name variable disappears. 

-----------------------------------------------------------------------------------
# Important
Parameter vs Argument

- def greet(name):

name is a parameter
because it's inside the function definition.

- def greet("Basmala")

"Basmala" is an argument
because it's the value you pass to the function.

So, I can say:
Parameter = the placeholder.
Argument = the actual value.


def greet(name):
           ↑
     Empty box (parameter)

Call:

greet("Basmala")
       ↑
Actual value (argument)
```

```python
# Multiple Parameters
A function can receive more than one parameter.

Syntax:
def function_name(parameter1, parameter2):
    # code

Python matches the arguments by their order.
```

```python
def introduce(name, age):
    print(f"My name is {name} and my age is {age}")


introduce("Basmala", 19)    
```

```python
# Look here:
# If I made a mistake and swapped the parameters 

def introduce(name, age):
    print(f"My name is {name} and my age is {age}")


introduce(19, "Basmala")    

# Python doesn't know that's wrong
# it simply assigns values based on their position.
```

```python
# return
This is one of the most important concepts in Python.

Let's compare between return and print
```

```python
def add(a, b):
    print(a + b)

add(4, 6)
```

```python
def add(a, b):
    return(a + b)

result = add(4, 6)    
print(result)
```

```python
Why is return powerful?

Imagine you want to calculate:

(5 + 3) × 10

If your function uses return:

answer = add(5, 3)

print(answer * 10)

Look at the next code:
```

```python
def add(a, b):
    return(a+b) 

answer = add(5, 3)
print(answer * 10)
```

```python
# Default Parameters
Sometimes we want a function to have a default value.
```

```python
def greet(name = "Guest"):
    print(f"Hello {name}")

greet()    
```

```python
def greet(name = "Guest"):
    print(f"Hello {name}")

greet("Basmala")  
```

```python
# Variable Scope

<<Local Variables>>
Variables created inside a function exist only inside that function

_Example in the nexr cell_
```

```python
def greet():
    message = "Hello"

    print(message)

greet()
```

```python
def greet():
    message = "Hello"

greet()

print(message)
```

```python
What happened?

Because message only exists inside greet().
When the function finishes, the variable disappears.
```

```python
<<Global Variables>>
A variable created outside every function is called a global variable.
```
