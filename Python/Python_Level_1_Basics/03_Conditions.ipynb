```python
# What if there are more than 2 possibilities?
That's where elif comes in 

if condition1:
    ...
elif condition2:
    ...
else:
    ...

---> Python checks the conditions from top to bottom. As soon as one condition is True, it executes that block and skips the rest
```

```python
grade = int(input("Enter your grade: "))

if grade >= 90:
    print("Grade: A")
elif grade >= 80:
    print("Grade: B")
elif grade >= 70:
    print("Grade: C")
elif grade >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
# Nested Conditions
A nested condition means writing an if statement inside another if statement

Syntax:
    if condition1:
     if condition2:
        # code

         not 

     if condition1:
    print("Level 1")

    if condition2:
        print("Level 2")    
```

```python
age = int(input("Enter your age: "))
has_id = input("Do you have an ID? (yes/no): ")

if age >= 18: 
    if has_id == "yes":
        print("Access granted.")
    else:
        print("You need an ID.")
else:
    print("You must be at least 18.")
```

```python
# Sometimes nested if statements can be simplified
Instead of:
if age >= 18:
    if has_id == "yes":
        print("Access granted.")
        
we can write:
if age >= 18 and has_id == "yes":
    print("Access granted.")
    
This is shorter and easier to read
```
