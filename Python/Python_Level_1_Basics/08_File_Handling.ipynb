```python
# Why Do We Need Files?

Imagine you're building:
- A student management system
- A BMI calculator
- A to-do list
- A notes app

If the program forgets everything after closing, it's not very useful.
Instead, we save the data in a file.

# File handling means:
Reading data from a file.
Writing data to a file.
```

```python
# to open a file 
- open()

Basic syntax:
file = open("notes.txt", "r")

notes.txt ---> This is the file name 
"r" ---> This is the mode, it tells py what I want to do with the file
```

```python
# File Modes

"r" (read)
Read a file

"w" (write)
Write to a file (creates a new file or deletes old content)

"a" (append)
Append (add to the end of the file)

"x" (create)
Create a new file (fails if it already exists)
```

```python
# Reading a File ("r")

Suppose notes.txt contains:
Python
Machine Learning
Data Science

# Code:
file = open("notes.txt", "r")
content = file.read()
print(content)

# Output:

Python
Machine Learning
Data Science
```

```python
# What Does read() Do?

read() reads everything inside the file and returns it as one string.

# So after:

content = file.read()
content is actually:
"Python\nMachine Learning\nData Science"

# Notice:
The whole file becomes one string.
```

```python
# Very Important: Close the File

After finishing, always write:
       <file.close()>

# Complete example:

file = open("notes.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
# The Best Way to Open Files: with open(...)

So far, I've written:

file = open("notes.txt", "r")
content = file.read()
print(content)
file.close()

<This works>
    
# But what if an error happens before file.close()?
The file might stay open

Python gives us a safer way:
with open("notes.txt", "r") as file:
    
    content = file.read()
    print(content)
    
Notice something?
There is no file.close().
Why?
Because when the with block finishes, Python automatically closes the file.
```

```python
# How Does It Work?

with open("notes.txt", "r") as file:
Let's break it down:

with → tells Python to manage the file for you.
                                      
open() → opens the file.
                                      
"r" → read mode.
                                      
as file → store the opened file in a variable called file.
                                      
Everything inside the indented block can use the file.
    
When the block ends:
print("Finished!")
The file is already closed automatically.
```

```python
with open("about_me.txt", "r") as file:
 content = file.read()
 print(content)
```

```python
with open("about_me.txt", "w") as file:
    file.write("Knock knock, who is knocking?")
```

```python
with open("about_me.txt", "a") as file:
    file.write("\n This is the best engineer ever!")
```

```python
# Final Summary for File Handling

open()
Open a file

"r"
Read a file

"w"
Write (overwrite) a file

"a"
Append to a file

read()
Read the whole file

readline()
Read one line

readlines()
Read all lines into a list

write()
Write text to a file

close()
Close the file

with open()
Open and close the file automatically (recommended)
```
