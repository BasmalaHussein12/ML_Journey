```python
1) Arithmetic operators:
+ ---> Addition
5 + 3

- ---> Subtraction
5 - 3

* ---> Multiplication
5 * 3

/ ---> Division
5 / 3

// ---> Floor Division
5 // 3

% ---> Modulus (Remainder)
5 % 3 

** ---> Power
5 ** 3
```

```python
a = 10
b = 3

print("Addition:" , a+b )
print("Substraction:" , a-b )
print("Multiplication:" , a*b )
print("Division:" , a/b )
print("Floor division:" , a//b )
print("Reminder:" , a%b )
print("Power:" , a**b )
```

```python
# Notes from the previous code: 
- When we use division /, the answer always returns a float
- When we use floor division //, it returns the whole-number quotient, discarding the decimal part
- Modules % are very useful to check if a number is even or odd (figure it out in the next example) 
```

```python
anounmous_number= int(input("Enter here your number: "))

if anounmous_number % 2 == 0:
    print("Your number is even!")

else:
    print("Your number is odd :(")
```

```python
# compression operations
Comparison operators compare two values and always return a Boolean (True or False).

(==) --> Equal to
5 == 5

(!=) --> Not equal to
5 != 3

(>) -->Greater than
5 > 3

(<) --> Less than
5 < 3

(>=) --> Greater than or equal to
5 >= 5

(<=) --> Less than or equal to
5 <= 3
```

```python
x = 10
y = 5

print("Equal:", x == y)
print("Not Equal:", x != y)
print("Greater Than:", x > y)
print("Less Than:", x < y)
print("Greater Than or Equal:", x >= y)
print("Less Than or Equal:", x <= y)
```

```python
# Important note!!
- = means assign a value to a variable
- == means compare two values
```

```python
age = 20

print(age == 20)
print(age > 18)
print(age < 18)
```

```python
#Comparing String 
name = "Basmala"

print(name == "Basmala")
print(name == "Ali")
```

```python
# Logical operators 
and ---> Both conditions must be True
or ---> At least one condition must be True
not ---> Reverses the Boolean value
```

```python
#and 
age = 19
has_id = True

print(age >= 18 and has_id)

print(10 > 5 and 3 > 7)
```

```python
# or 
age = 19
has_id = True

print(10 > 5 or 3 > 7)

print(False or False)
```

```python
# not
is_student = True

print(not is_student)

print(not False)
```
