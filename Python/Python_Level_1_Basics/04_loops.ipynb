```python
# Loops
In this notebook, I will learn:
- Why loops are important
- for loop
- while loop
- range()
- break
- continue
- Nested loops
```

```python
Imagine printing:
Hello
Hello
Hello
Hello
Hello

Without loops:
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")

That's repetitive.
```

```python
# for 
for variable in iterable:
    # code

The variable takes a new value each time the loop runs    
```

```python
# range
for i in range(5):
    print(i)
```

```python
In the previous example, it doesn't print 5 
because ---> Start = 0
        ---> Stop = 5 (not included)

** We can also control start and stop
```

```python
for i in range(3, 8):
    print(i)
```

```python
# ** In addition, we can control the step
for i in range(2, 11, 2):
    print(i)
```

```python
# Looping Through a String
# A string is a sequence of characters, so we can loop through it
name = "Basmala"

for letter in name:
    print(letter)
```

```python
# Looping Through a List
Loops become even more powerful with lists
```

```python
fruits = ["Apple", "Banana", "Orange"]

for fruit in fruits:
    print(fruit)
```

```python
# The break Statement
for i in range(1, 11):
    if i == 6:
        break
# When i becomes 6, the loop ends immediately
    print(i)
```

```python
# The continue Statement
continue is different
It skips the current iteration and moves to the next one
```

```python
for i in range(1, 6):
    if i == 3:
        continue

    print(i)
```

```python
# The while Loop
Unlike a for loop, a while loop repeats as long as a condition is True

Syntax
while condition:
    # code
Python checks the condition before each iteration
```

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
# Nested Loops
A nested loop is simply a loop inside another loop

Syntax
for i in range(...):
    for j in range(...):
        # code
        
The inner loop runs completely every time the outer loop runs once
```

```python
for i in range(3):
    for j in range(2):
        print(f"i = {i}, j = {j}")
```

```python
# Printing Patterns
Nested loops are often used to print patterns
```

```python
for i in range(3):
    for j in range(4):
        print("*", end="")
    print()
```

```python
# When should I use for and when should I use while?
for
You know how many times to repeat
Loop through a list, string, or range

while
You don't know how many times
Repeat until a condition changes
```

```python
Final Project — Number Guessing Game:

1)Store a secret number (for example, 7)
2)Ask the user to guess it
3)If the guess is wrong, ask again
4)When the guess is correct, print:
# Congratulations! You guessed the number

Hint: Think about which loop (for or while) is more suitable
```

```python
secret_number = 7
print("Hello!!")
print("Please Guess a number and write it here: ")

guessed= int(input())
while guessed != secret_number:
  print("Wrong guess. Try again!")
  guessed = int(input("Your guess: "))
    
print("Congratulations! You guessed the number")
```
