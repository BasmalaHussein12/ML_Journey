```python
# Lists []
Stores multiple values in one variable

Instead of:

fruit1 = "Apple"
fruit2 = "Banana"
fruit3 = "Orange"
fruit4 = "Mango"

we can create a list:
fruits = ["Apple", "Banana", "Orange", "Mango"]

<< A list can store any type of data and different data types >>
numbers = [10, 20, 30, 40]
data = ["Basmala", 19, 3.34, True]
```

```python
# Accessing Elements (Indexing)
Every element has an index.

Value :  Apple   Banana   Orange   Mango
Index :    0        1        2        3

So, I noticed that python starting with 0, not 1
```

```python
fruits = ["Apple", "Banana", "Orange", "Mango"]

print(fruits[0])
```

```python
# Negative Indexing
Python can count from the end.

Apple   Banana   Orange   Mango
 0         1        2        3

-4       -3       -2       -1

<< LoL it's very easy >>
```

```python
fruits = ["Apple", "Banana", "Orange", "Mango"]

print(fruits[-2])
```

```python
# Modifying Lists
Unlike strings, lists are mutable.
That means you can change them.
```

```python
fruits = ["Apple", "Banana", "Orange"]

fruits[1] = "Mango"

print(fruits)
```

```python
len()
To know how many elements a list contains:
# notice that starts the count from 1
```

```python
numbers = [5, 10, 15, 20]

print(len(numbers))
```

```python
# Adding Elements ---> append()
The append() method adds an element to the end of the list.
```

```python
fruits = ["Apple", "Banana"]

fruits.append("Orange")

print(fruits)
```

```python
# insert()
The insert() method adds an element at a specific index.
    
Syntax
list_name.insert(index, value)
```

```python
fruits = ["Apple", "Banana"]

fruits.insert(1,"Orange")

print(fruits)
# Index 1 is where the new element is inserted.
# The existing elements shift to the right.
```

```python
# Removing Elements
The remove() method removes an element by its value.
<< If the value doesn't exist, Python raises a ValueError >>
```

```python
fruits = ["Apple", "Banana", "Orange"]

fruits.remove("Banana")

print(fruits)
```

```python
# pop()
The pop() method removes an element by its index

fruits.pop()
<< Python removes the last element >>
```

```python
fruits = ["Apple", "Banana", "Orange"]

fruits.pop(1)

print(fruits)
```

```python
fruits = ["Apple", "Banana", "Orange"]

fruits.pop()

print(fruits)
```

```python
# I've noticed that:
remove("Banana")   → Removes by VALUE (If there is no value, it raises an error)
pop(1)             → Removes by INDEX (If there is no index, it removes the last element)
```

```python
# Sorting a List
Use sort() to sort the list in ascending order.
```

```python
numbers = [8, 3, 1, 6]

numbers.sort()

print(numbers)
```

```python
# If I want a descending order
numbers = [8, 3, 1, 6]

numbers.sort(reverse=True)

print(numbers)
```

```python
# Reversing a List
Use reverse() to reverse the current order of the list.
```

```python
numbers = [1, 2, 3, 4]

numbers.reverse()

print(numbers)
```

```python
# Important thing I've noticed:

<< numbers.sort(reverse=True) >>
means:
Sort from largest to smallest.

While:

<< numbers.reverse() >>
means:
Just reverse the current order, without sorting.
```

```python
And its common methods: 

append()
Add an element to the end

insert()
Add an element at a specific index

remove()
Remove an element by value

pop()
Remove an element by index (or last element if no index is given)

sort()
Sort the list

reverse()
Reverse the current order

len()
Return the number of elements
```

```python
# Tuples ()

What is a Tuple?
A tuple is a collection of multiple items.

<< It is very similar to a list >>

The biggest difference is: Tuples are immutable

-----> That means you cannot change, add, or remove elements after creating the tuple.

We create a tuple with ()

We can also store numbers or different data types.
```

```python
colors = ("Red", "Green", "Blue")
numbers = (10, 20, 30, 40)
data = ("Basmala", 19, 3.34, True)
```

```python
# Accessing elements  
Just like lists[], tuples use indexing

Negative indexing also works.
```

```python
colors = ("Red", "Green", "Blue")

print(colors[1])
```

```python
colors = ("Red", "Green", "Blue")

print(colors[-2])
```

```python
To know the number of elements, we also use len()
```

```python
colors = ("Red", "Green", "Blue")

print(len(colors))
```

```python
# Sets {}
What is a Set?
A set is a collection of unique elements.

The keyword here is: << Unique >>

If you add duplicate values, Python automatically removes them.

Let's create a set with {}
```

```python
fruits = {"Apple", "Banana", "Orange"}
```

```python
numbers = {1, 2, 3, 2, 1, 5}

print(numbers)
```

```python
By creating this example, I've noticed that the duplicate 1 and 2 disappeared automatically.
```

```python
# Sets Are Unordered
Unlike lists and tuples, sets do not preserve the order of elements.
```

```python
colors = {"Red", "Green", "Blue"}

print(colors)
```

```python
The output might be:

{'Blue', 'Red', 'Green'}

or

{'Green', 'Blue', 'Red'}

The order is not guaranteed.
```

```python
Sets Are Not Indexed

This is invalid:
```

```python
colors = {"Red", "Green", "Blue"}

print(colors[0])
```

```python
TypeError: 'set' object is not subscriptable

Because sets don't have indexes.
(mande2y because it's not ordered)
```

```python
For adding elements use add() and for removing elements use remove()
(If the value doesn't exist, Python raises a KeyError)

if I want no error, I can use discord indstead of remove
```

```python
colors = {"Red", "Green"}

colors.add("Blue")

print(colors)
```

```python
colors = {"Red", "Green"}

colors.remove("Red")

print(colors)
```

```python
colors = {"Red", "Green"}

colors.remove("Blue")

print(colors)
```

```python
colors = {"Red", "Green"}

colors.discard("Blue")

print(colors)
```

```python
We also use len() for returning the number of elements
```

```python
numbers = {10, 20, 30}

print(len(numbers))
```

```python
# Summary for Sets: {}
While I'm learning about sets, I've noticed that sets are:

- mutable
- can store different types of data

but not:
- ordered 
- indexed
- duplicates 
```

```python
<< Common Methods >>

add()
Add an element

remove()
Remove an element (raises an error if not found)

discard()
Remove an element (no error if not found)

len()
Return the number of elements
```

```python
List → Ordered and editable.
Tuple → Ordered but fixed.
Set → Unique values only, no order.
```

```python
# Dictionaries {}
are one of the most important data structures in Python. You'll use them constantly in:
- Machine Learning
- APIs
- JSON files
- Web development
- Data Science

What is a Dictionary?
A dictionary stores data as key-value pairs.
    
Think of a real dictionary:
Word  → Meaning

Apple → A fruit
Book  → A collection of pages
    
Python dictionaries work the same way.
    
Creating a Dictionary
Use curly braces {}.
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}

here: 

"name"  → key
"Basmala" → value

"age" → key
19 → value
```

```python
# Accessing values
to access an value, we will use the key 
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}

print(student["name"])
```

```python
Also, to modify or add an value, we will use the key 
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}

student["age"] = 20

print(student)
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}

student["GPA"] = 3.333

print(student)
```

```python
# removing items by pop()

student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}

student.pop("age")

print(student)
```

```python
len()
Return the number of key-value pairs
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}
print(len(student))
```

```python
# getting all values 
.values()
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}
print(student.values())
```

```python
# getting all keys
.keys()
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}
print(student.keys())
```

```python
If I want to print every key and its value, I will looping

for key, value in student.items():
    print(key, ":", value)
```

```python
student = {
    "name": "Basmala",
    "age": 19,
    "major": "Software Industry"
}

for key, value in student.items():
    print(key, ":", value)
```

```python
# Summary for dictionaries: { key : value }
While I'm learning about dicts, I've noticed that dicts are:

- mutable
- ordered 
- duplicate values not keys 
- can store different types of data

but not:
- indexed (use keys instead)
```

```python
# Method

keys()
Return all keys

values()
Return all values

items()
Return key-value pairs

pop()
Remove a key-value pair

len()
Return the number of pairs
```
