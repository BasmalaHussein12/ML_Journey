```python
import numpy as np

X = np.array([[1],[2],[3],[4]])
w = np.array([2])

print("="*50)
print("DIFFERENT WAYS TO ACCESS X[1]")
print("="*50)

# Method 1: Indexing (returns 1D)
print(f"X[1]           = {X[1]}, shape: {X[1].shape}")  
# [2], shape: (1,)

# Method 2: Slicing (returns 2D)
print(f"X[1:2]         = {X[1:2]}, shape: {X[1:2].shape}")  
# [[2]], shape: (1, 1)

# Method 3: Reshape (can change shape)
print(f"X[1].reshape(1,1) = {X[1].reshape(1,1)}, shape: {X[1].reshape(1,1).shape}")  
# [[2]], shape: (1, 1)

print("\n" + "="*50)
print("DIFFERENT WAYS TO DO THE SAME CALCULATION")
print("="*50)

# Method 1: Both 1D arrays (returns scalar)
c1 = np.dot(X[1], w)
print(f"dot(X[1], w) = {c1}, shape: {c1.shape}")  
# 4, shape: ()

# Method 2: Both 2D arrays (returns 2D)
c2 = np.dot(X[1:2], w.reshape(1,1))
print(f"dot(X[1:2], w.reshape(1,1)) = {c2}, shape: {c2.shape}")  
# [[4]], shape: (1, 1)

# Method 3: Matrix multiplication (2D only)
c3 = np.matmul(X[1:2], w.reshape(1,1))
print(f"matmul(X[1:2], w.reshape(1,1)) = {c3}, shape: {c3.shape}")  
# [[4]], shape: (1, 1)
```

## Now, I will compare between for `dot function` and `for loop`:

```python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

## Now, comparing the time 

```python
import time 

np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

- So, vectorization provides a large speed up in this example. This is because NumPy makes better use of available data parallelism in the underlying hardware. GPU's and modern CPU's implement Single Instruction, Multiple Data (SIMD) pipelines allowing multiple operations to be issued in parallel. This is critical in Machine Learning where the data sets are often very large.

```python
print("="*50)
print("DIFFERENT WAYS TO CREATE A COLUMN VECTOR")
print("="*50)

# Method 1: All in one line
a1 = np.array([[5], [4], [3]])
print(f"Method 1: {a1}")
print(f"Shape: {a1.shape}\n")

# Method 2: Multi-line (more readable)
a2 = np.array([[5],
               [4],
               [3]])
print(f"Method 2: {a2}")
print(f"Shape: {a2.shape}\n")

# Method 3: Using reshape
a3 = np.array([5, 4, 3]).reshape(3, 1)
print(f"Method 3: {a3}")
print(f"Shape: {a3.shape}\n")

# Method 4: Using reshape with -1 (auto-calculate)
a4 = np.array([5, 4, 3]).reshape(-1, 1)
print(f"Method 4: {a4}")
print(f"Shape: {a4.shape}\n")

# Method 5: Using np.newaxis
a5 = np.array([5, 4, 3])[:, np.newaxis]
print(f"Method 5: {a5}")
print(f"Shape: {a5.shape}\n")

print("="*50)
print("ALL METHODS PRODUCE THE SAME RESULT!")
print("="*50)
```

```python
# X is typically a 2D array (samples, features)
X = np.array([[1, 2, 3],    # Sample 1: features [1,2,3]
              [4, 5, 6]])   # Sample 2: features [4,5,6]
# Shape: (2, 3) → 2 samples, 3 features each

# w is a column vector (features, 1) or 1D array
w = np.array([[0.5], 
              [0.5], 
              [0.5]])  # Shape: (3, 1)

# OR
w = np.array([0.5, 0.5, 0.5])  # Shape: (3,)

# Predictions: X @ w (if using column vector)
pred = np.dot(X, w)  # Shape: (2, 1) - column vector

# OR
pred = np.dot(X, w)  # Shape: (2,) - 1D array
```

```python
# Shape: (number of samples, number of features)
X = np.array([[1.0, 2.0, 3.0],    # Sample 1: 3 features
              [4.0, 5.0, 6.0],    # Sample 2: 3 features
              [7.0, 8.0, 9.0]])   # Sample 3: 3 features
# Shape: (3, 3) → 3 samples, 3 features each

# Access sample 1:
X[0]  # Shape: (3,) - 1D array

# Access feature 1 of sample 1:
X[0, 0]  # Shape: () - scalar
```

```python
# Option 1: 1D array (most common)
w = np.array([0.5, 0.5, 0.5])  # Shape: (3,)
b = 10.0  # Shape: () - scalar

# Option 2: Column vector (less common)
w_col = np.array([[0.5],
                  [0.5],
                  [0.5]])  # Shape: (3, 1)
```

```python
# With 1D w
X = np.array([[1, 2, 3], [4, 5, 6]])  # Shape: (2, 3)
w = np.array([0.5, 0.5, 0.5])         # Shape: (3,)
b = 10

pred = np.dot(X, w) + b  # Shape: (2,) - 1D array
print(pred)  # [13.0, 17.5]

# With column vector w
w_col = w.reshape(-1, 1)  # Shape: (3, 1)
pred_col = np.dot(X, w_col) + b  # Shape: (2, 1) - 2D array
print(pred_col)  # [[13.0], [17.5]]
```

```python
#vector 2-D slicing operations
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#access 5 consecutive elements (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

#access 5 consecutive elements (start:stop:step) in two rows
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# access all elements
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# access all elements in one row (very common usage)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# same as
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
