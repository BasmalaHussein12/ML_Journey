# Linear Regression — Cost Function

## 1. Goal of Linear Regression

The goal of linear regression is to find a model:

$$
f_{w,b}(x) = wx + b
$$

that can accurately predict the target value $y$ from an input $x$.

Where:

- $x$: input feature
- $y$: actual/target value
- $w$: weight/parameter
- $b$: bias/parameter
- $f_{w,b}(x)$: prediction made by the model

The goal is to find the values of $w$ and $b$ that make the predictions as accurate as possible.

## 2. Model Prediction

For each training example $i$, the model makes a prediction:

$$
f_{w,b}(x^{(i)}) = wx^{(i)} + b
$$

For example, if:

$$
w = 2,\quad b = 1,\quad x = 3
$$

then:

$$
f_{w,b}(3) = 2(3) + 1 = 7
$$

So, the model predicts:

$$
\hat{y} = 7
$$

## 3. Measuring the Error

To determine how well the model is performing, we compare its prediction with the actual target value.

The error for one training example is:

$$
f_{w,b}(x^{(i)}) - y^{(i)}
$$

We square the error:

$$
\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2
$$

Squaring prevents positive and negative errors from cancelling each other out and gives larger errors a larger penalty.

## 4. Cost Function

The cost function measures how well the model fits the training data.

For linear regression:

$$
J(w,b)
=
\frac{1}{2m}
\sum_{i=0}^{m-1}
\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2
$$

Where:

- $J(w,b)$: cost of the model
- $m$: number of training examples
- $f_{w,b}(x^{(i)})$: prediction for example $i$
- $y^{(i)}$: actual target value

### Main idea

$$
\boxed{\text{Lower Cost} = \text{Better Model}}
$$

Therefore, our goal is:

$$
\boxed{\min_{w,b} J(w,b)}
$$

which means finding the values of $w$ and $b$ that minimize the cost.

````python
def compute_cost(x, y, w, b):
    """
    Computes the cost function for linear regression.
    """
    
    # Number of training examples
    m = x.shape[0]
    
    cost_sum = 0
    
    for i in range(m):
        
        # Model prediction
        f_wb = w * x[i] + b
        
        # Squared error
        cost = (f_wb - y[i]) ** 2
        
        # Add error to total
        cost_sum = cost_sum + cost
    
    # Compute final cost
    total_cost = (1 / (2 * m)) * cost_sum
    
    return total_cost
````

## 5. Understanding `compute_cost()`

The function `compute_cost()` does **not** find the best parameters.

It answers the question:

> How good is the model for these specific values of $w$ and $b$?

For example:

```python
compute_cost(x, y, w=100, b=50)



## 6. Finding the Optimal Parameters

Our actual goal is:

$$
\min_{w,b} J(w,b)
$$

We want to find the values of $w$ and $b$ that produce the smallest possible cost.

In the previous lab, we found that:

$$
b = 100
$$

provides an optimal solution.

Therefore, we can fix:

$$
b = 100
$$

and focus on finding the optimal value of $w$.

The model becomes:

$$
f_{w,100}(x) = wx + 100
$$

## 7. Using the Slider

The slider allows us to manually change the value of $w$.

For every value of $w$, the notebook calculates:

$$
J(w,100)
$$

We can then observe how the cost changes.

For example:

$$
w = 50 \rightarrow \text{high cost}
$$

$$
w = 100 \rightarrow \text{lower cost}
$$

$$
w = 150 \rightarrow \text{lower cost}
$$

$$
w = 200 \rightarrow \text{minimum cost}
$$

$$
w = 250 \rightarrow \text{higher cost}
$$

The value of $w$ that produces the smallest cost is the best value of $w$ for this example.

The slider is therefore a visualization tool that helps us understand the relationship between $w$ and the cost.

## 8. The Slider vs. Gradient Descent

The slider is **not** Gradient Descent.

With the slider, we manually try different values of $w$ and observe the corresponding cost.

In real machine learning, we do not want to manually test values.

Instead, we use **Gradient Descent** to automatically update $w$ and $b$ toward values that minimize the cost.

### Slider

$$
w \rightarrow J(w,b)
$$

We manually change $w$ and observe the cost.

### Gradient Descent

$$
w,b \rightarrow \text{update} \rightarrow w,b \rightarrow \text{lower cost}
$$

Gradient Descent automatically searches for better parameters.

# 9. The Big Picture

The complete process is:

$$
x
\rightarrow
f_{w,b}(x)
\rightarrow
\text{Prediction}
\rightarrow
\text{Error}
\rightarrow
\text{Cost}
\rightarrow
\text{Optimization}
$$

More specifically:

$$
f_{w,b}(x) = wx+b
$$

↓

Make predictions

↓

Compare predictions with actual values:

$$
f_{w,b}(x^{(i)}) - y^{(i)}
$$

↓

Calculate squared errors:

$$
\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2
$$

↓

Calculate the total cost:

$$
J(w,b)
$$

↓

Find the parameters that minimize the cost:

$$
\boxed{w^*,b^*}
$$

These optimal parameters produce the best-fitting linear regression model.

### Remember

> **The model makes predictions, the cost function measures how bad those predictions are, and optimization methods such as Gradient Descent find the parameters that minimize the cost.**
