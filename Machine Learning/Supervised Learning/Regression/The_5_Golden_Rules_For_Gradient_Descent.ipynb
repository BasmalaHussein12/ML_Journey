## [Rule 2: Gradients Must Get Smaller](#rule-2-gradients-must-get-smaller)

### The Rule

> **As you approach the minimum, gradients should approach ZERO.**


````python
def check_gradients(J_history, num_iters=10):
    """
    Check if gradients are decreasing properly.
    """
    print("\nGRADIENT CHECK:")
    print("-" * 50)
    print("Iteration |  Cost     | Gradient    | Status")
    print("-" * 50)
    
    # Print every 10% of iterations
    for i in range(0, len(J_history), max(1, len(J_history)//10)):
        if i == 0:
            continue
        cost = J_history[i]
        prev_cost = J_history[i-1]
        change = abs(cost - prev_cost)
        
        if change < 0.01:
            status = "CONVERGED"
        elif cost < prev_cost:
            status = "Improving"
        else:
            status = "DIVERGING!"
        
        print(f"{i:9d} | {cost:9.2f} | {change:11.4f} | {status}")
````

## [Rule 3: Parameters Must Stabilize](#rule-3-parameters-must-stabilize)

### The Rule

> **w and b should stop changing (or change very little) near the end.**


````python
def check_parameters(w_history, num_iters=10):
    """
    Check if parameters are converging.
    """
    print("\nPARAMETER CONVERGENCE CHECK:")
    print("-" * 70)
    print("Iteration |  w[0]    |  w[1]    |  w[2]    |  w[3]    |   b")
    print("-" * 70)
    
    # Show first, middle, and last
    indices = [0, len(w_history)//4, len(w_history)//2, 
               3*len(w_history)//4, len(w_history)-1]
    
    for idx in indices:
        if idx < len(w_history):
            w, b = w_history[idx]
            print(f"{idx:9d} | {w[0]:8.4f} | {w[1]:8.4f} | {w[2]:8.4f} | {w[3]:8.4f} | {b:8.4f}")
    
    # Check last 10% of values
    last_start = int(0.9 * len(w_history))
    last_ws = [w[0] for w, _ in w_history[last_start:]]
    last_bs = [b for _, b in w_history[last_start:]]
    
    w_change = np.std(last_ws)  # Standard deviation of changes
    b_change = np.std(last_bs)
    
    if w_change < 0.01 and b_change < 0.01:
        print("\nParameters have converged!")
    else:
        print(f"\nParameters still changing (w_std={w_change:.4f})")
        print("→ Need more iterations or smaller alpha")
````

## [Rule 4: Predictions Must Make Sense](#rule-4-predictions-must-make-sense)

### The Rule

> **The model should make predictions that make SENSE!**


````python
def check_predictions(X, y, w, b):
    """
    Check if predictions make sense.
    """
    predictions = np.dot(X, w) + b
    errors = predictions - y
    
    print("\nPREDICTION CHECK:")
    print("-" * 60)
    print("Sample | Predicted | Actual  | Error    | Status")
    print("-" * 60)
    
    for i in range(len(y)):
        error = errors[i]
        if abs(error) < 5:
            status = "Good"
        elif abs(error) < 50:
            status = "Okay"
        else:
            status = "Bad!"
        
        print(f"{i:6d} | {predictions[i]:8.2f} | {y[i]:7.2f} | {error:8.2f} | {status}")
    
    mae = np.mean(np.abs(errors))
    print(f"\nMean Absolute Error: {mae:.2f}")
    
    if mae < 10:
        print("Excellent predictions!")
    elif mae < 50:
        print("Reasonable predictions, could improve")
    else:
        print("Predictions need improvement!")
        print("→ Check if features are scaled")
        print("→ Try more iterations")
        print("→ Adjust learning rate")
````

## [Rule 5: Learning Curve Must Look Like Learning](#rule-5-learning-curve-must-look-like-learning)

### The Rule

> **The learning curve should look like a smooth exponential decay.**


### Learning Curve Patterns

**GOOD: Smooth Learning Curve**

```text
Cost
  ^
  | \
  |  \_
  |   \__
  |     \___
  |        \___
  |           \___
  +-----------------> Iterations

````python
# Case 1: Bouncing curve
alpha = 0.01  # Current alpha
alpha = alpha / 10  # New alpha = 0.001

# Case 2: Flat curve
alpha = 0.0001  # Current alpha
alpha = alpha * 10  # New alpha = 0.001

# Case 3: Increasing curve
alpha = 0.1  # Current alpha
alpha = alpha / 100  # New alpha = 0.001
````
