# Robust Scaling

## 1. What is Robust Scaling?

**Robust Scaling** is a Feature Scaling technique designed to be less sensitive to outliers.

Unlike Standardization and Min-Max Scaling, Robust Scaling uses the:

- Median
- Interquartile Range (IQR)

This makes it useful when the dataset contains significant outliers.

### Formula

$$
x' = \frac{x-\text{median}}{IQR}
$$

Where:

$$
IQR = Q_3-Q_1
$$

- $Q_1$ = First Quartile (25th percentile)
- $Q_3$ = Third Quartile (75th percentile)

## 2. What is the Median?

The **median** is the middle value of an ordered dataset.

For example:

```text
[10, 12, 15, 18, 20]

the median = 15


## 3. What is IQR?

**IQR** stands for **Interquartile Range**.

It measures the spread of the middle 50% of the data.

The formula is:

$$
IQR = Q_3-Q_1
$$

Where:

- $Q_1$ = 25th percentile
- $Q_3$ = 75th percentile

The IQR is useful because it is less affected by extreme values.

## 4. Why Do We Use Robust Scaling?

Consider this dataset:

```text
[10, 12, 13, 15, 18, 20, 1000]

The value 1000 is an extreme outlier.

Scaling methods that depend on the mean, standard deviation, minimum, or maximum can be strongly affected by this value.

Robust Scaling uses the median and IQR instead.

````python
from sklearn.datasets import load_diabetes

data = load_diabetes(as_frame=True)

df = data.frame
df
````

````python
df.describe().T
````

````python
x = df.drop("target", axis=1)
y = df["target"]
````

````python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

x_scaled = scaler.fit_transform(x)
x_scaled
````

````python
scaler.center_
````

`center_` contains the median calculated for each feature.

````python
scaler.scale_
````

`scale_` contains the scaling factor, which is based on the IQR of each feature.

## 5. Understanding the Result

After Robust Scaling:

- The median of each feature becomes approximately `0`.
- The IQR is scaled to approximately `1`.

The data is centered around the median rather than the mean.

## 9. When Should We Use RobustScaler?

RobustScaler is useful when:

- The dataset contains outliers.
- The features have very different scales.
- You do not want extreme values to strongly affect the scaling.
- The median and IQR are more representative than the mean and standard deviation.


## 10. Key Takeaways

- **Robust Scaling** is a Feature Scaling technique.
- It is designed to be less sensitive to outliers.
- It uses the **Median** and **IQR**.
- The formula is:

$$
x' = \frac{x-\text{median}}{IQR}
$$

- `RobustScaler()` performs Robust Scaling.
- `fit()` calculates the median and IQR of each feature.
- `transform()` applies the scaling.
- `fit_transform()` performs both operations.
- The median becomes approximately `0`.
- The IQR becomes approximately `1`.
- There is no fixed output range.
- The scaler should be fitted only on training data to avoid Data Leakage.
