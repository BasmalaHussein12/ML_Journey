# Normalization

## 1. What is Normalization?

**Normalization** is a data preprocessing technique that scales each sample (row) so that it has a unit norm.

Unlike Standardization and Min-Max Scaling, Normalization works on **rows/samples**, not on individual features/columns.

The most common type is **L2 Normalization**.

---

## 2. Why Do We Use Normalization?

Normalization is useful when the **direction or proportions of a vector** are more important than its absolute magnitude.

For example, consider:

[3, 4]

The length of this vector is:

$$
\sqrt{3^2 + 4^2} = 5
$$

After L2 Normalization:

$$
[3,4] \rightarrow [0.6,0.8]
$$

3/5 = 0.6, 4/5 = 0.8 

The direction of the vector stays the same, but its length becomes 1.

---

## 3. L2 Normalization

The L2 norm of a vector is:

$$
||x||_2 = \sqrt{x_1^2+x_2^2+\dots+x_n^2}
$$

Each value is then divided by the L2 norm:

$$
x_i' = \frac{x_i}{||x||_2}
$$

For example:

$$
x=[3,4]
$$

The L2 norm is:

$$
||x||_2 = \sqrt{3^2+4^2}=5
$$

Therefore:

$$
x'=[3/5,4/5]
$$

$$
x'=[0.6,0.8]
$$

## 4. Normalizer in Scikit-learn

Scikit-learn provides `Normalizer` to perform normalization.

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

data = load_breast_cancer(as_frame=True)

df = data.frame
df
```

```python
df.describe().T
```

```python
x = df.drop("target", axis=1)
y = df["target"]
```

At this point, no normalization has happened yet.

However, unlike `StandardScaler` and `MinMaxScaler`, `Normalizer` does not need to learn statistics such as the mean, minimum, or maximum.

---

## 5. Applying Normalization

We can normalize the data using:

```python
from sklearn.preprocessing import Normalizer

scaler = Normalizer()

x_normalized = scaler.fit_transform(x)
```

`transform()` normalizes each row independently.

For each row, the values are divided by the row's norm.

---

```python
import numpy as np

x_ex = np.array([[3,4]])

scaler = Normalizer()

x_normalized = scaler.fit_transform(x_ex)
x_normalized
```

# To check the Norm

```python
L = np.linalg.norm(x_normalized, axis=1)
L
```

This confirms that the row has been normalized to unit length.

axis = 1, it calculates the rows 

axis = 0, it calculates the columns
