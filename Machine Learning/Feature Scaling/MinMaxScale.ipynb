# Min-Max Scaling

## 1. What is Min-Max Scaling?

**Min-Max Scaling** is a Feature Scaling technique used to transform numerical features into a specific range.

The most common range is:

$$
[0,1]
$$

Unlike Standardization, Min-Max Scaling does **not** make the mean equal to 0 or the standard deviation equal to 1.

### Formula

$$
x' = \frac{x-x_{min}}{x_{max}-x_{min}}
$$

Where:

- $x$ = original value
- $x_{min}$ = minimum value of the feature
- $x_{max}$ = maximum value of the feature
- $x'$ = scaled value

After scaling:

- Minimum value → 0
- Maximum value → 1
- Other values → between 0 and 1

## 3. Why Do We Use Min-Max Scaling?

Different features can have very different scales.

For example:

- Age → 18–60
- Salary → 5,000–100,000
- Area → 50–10,000

Min-Max Scaling transforms them into a common range, usually:

[0, 1]

This makes the features comparable in scale.

## 9. Applying Min-Max Scaling

We can apply Min-Max Scaling using `fit_transform()`.

````python
x_minmax = scaler.fit_transform(x)
````

`fit_transform()` combines two operations.

### `fit()`

The scaler finds the:

- Minimum value
- Maximum value

of **each feature independently**.

### `transform()`

The scaler uses these values to transform each feature using:

$$
x' = \frac{x-x_{min}}{x_{max}-x_{min}}
$$

## 11. Changing the Feature Range

The default range of `MinMaxScaler` is:

[0, 1]

However, we can choose another range using `feature_range`.

````python
scaler = MinMaxScaler(feature_range=(-1,1))
x_minmax = scaler.fit_transform(x)
````

## 14. Min-Max Scaling and Outliers

Min-Max Scaling is sensitive to **outliers**.

For example:

10, 12, 15, 18, 20, 1000

Here, `1000` is an extreme value.

Since Min-Max Scaling uses the minimum and maximum values, `1000` becomes the maximum.

This can compress most of the other values into a small part of the [0, 1] range.

## 15. Standardization vs Min-Max Scaling

### Standardization

Uses:

$$
z = \frac{x-\mu}{\sigma}
$$

Based on:

- Mean
- Standard Deviation

Result:

- Mean ≈ 0
- Standard Deviation ≈ 1
- No fixed range

Using:

`StandardScaler()`

### Min-Max Scaling

Uses:

$$
x' = \frac{x-x_{min}}{x_{max}-x_{min}}
$$

Based on:

- Minimum
- Maximum

Result:

- Minimum = 0
- Maximum = 1
- Usually between 0 and 1

Using:

`MinMaxScaler()`
