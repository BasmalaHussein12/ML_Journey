# Apply Absolute Maximum Scaling

```python
import numpy as np
import pandas as pd

df = pd.read_csv("SampleFile.csv")
df = df.select_dtypes(include=np.number)
df.head()
```

```python
max_abs = np.max(np.abs(df), axis = 0)

scaled = df / max_abs

scaled.head()
```

# Min-Max Scaling

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)
scaled_data
```

```python
scaled_named_col = pd.DataFrame(scaled_data, columns=df.columns)

scaled_named_col.head()
```

# Normalization (Vector Normalization)

```python
from sklearn.preprocessing import Normalizer

scaler_normalize = Normalizer()

normalized_data = scaler_normalize.fit_transform(df)
normalized_data
```

```python
normalized_data_named_col = pd.DataFrame(normalized_data, columns=df.columns)
normalized_data_named_col.head()
```

# Standardization

```python
from sklearn.preprocessing import StandardScaler

scaler_standardization = StandardScaler()

scaler_standardization_scaled = scaler_standardization.fit_transform(df)
scaler_standardization_scaled
```

```python
scaler_standardization_scaled_named_col = pd.DataFrame(scaler_standardization_scaled, columns=df.columns)
scaler_standardization_scaled_named_col.head()
```

# Robust Scaling

```python
from sklearn.preprocessing import RobustScaler

scaler_RS = RobustScaler()

scaler_RS_scaled = scaler_RS.fit_transform(df)
scaler_RS_scaled
```

```python
scaler_RS_scaled_named_col = pd.DataFrame(scaler_RS_scaled, columns=df.columns)
scaler_RS_scaled_named_col.head()
```
