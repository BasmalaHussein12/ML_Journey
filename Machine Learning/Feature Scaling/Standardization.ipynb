# Standardization

## What is Standardization?

**Standardization** is a Feature Scaling technique used to transform numerical features so that they have approximately:

- Mean = 0
- Standard Deviation = 1

It is also called **Z-score Standardization**.

### Formula

$$
z = \frac{x - \mu}{\sigma}
$$

Where:

- $x$ = original value
- $\mu$ = mean of the feature
- $\sigma$ = standard deviation
- $z$ = standardized value

## Why Do We Use Standardization?

Different features can have very different scales.

For example:

- Age → 20–60
- Salary → 5,000–100,000
- Area → 100–10,000

Machine Learning algorithms can be affected by these differences in scale.

Standardization puts numerical features on a **comparable scale**.

## StandardScaler

`StandardScaler` is a Scikit-learn tool used to perform standardization.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

At this point, scaling has **not happened yet**.

We have only created a `StandardScaler` object.

## Applying Standardization

We can apply standardization using `fit_transform()`.

### What does `fit_transform()` do?

`fit_transform()` combines two operations:

#### 1. `fit()`

The scaler calculates the required statistics for **each feature independently**:

- Mean
- Standard deviation

#### 2. `transform()`

The scaler uses these statistics to transform the values using:

$$
z = \frac{x - \mu}{\sigma}
$$

Therefore:

```python
scaler.fit_transform(X)

scaler.fit(X)
X_scaled = scaler.transform(X)

X.head()

X_scaled[:5]

The values have changed, but the original information has not been removed.

Only the **scale of the features** has changed.

X_scaled.mean(axis=0)

The mean of each scaled feature should be approximately **0**.

The mean of each scaled feature should be approximately **0**.

X_scaled.std(axis=0)

The standard deviation of each scaled feature should be approximately **1**.

This confirms that Standardization worked correctly.

## Checking the Shape

The shape remains:

**(569, 30)**

Scaling does not change:

- Number of samples
- Number of features

It only changes the values.

## Key Takeaways

- **Standardization** is a Feature Scaling technique.
- It transforms features to have approximately:
  - Mean = 0
  - Standard Deviation = 1
- The formula is:

$$
z = \frac{x - \mu}{\sigma}
$$

- `StandardScaler()` performs standardization.
- `fit()` calculates the mean and standard deviation.
- `transform()` applies the transformation.
- `fit_transform()` performs both operations.
- Each feature is scaled independently.
- Scaling is normally applied to `X`.
- The target `y` is usually not scaled.
- Scaling does not change the shape of the dataset.
- It only changes the numerical scale of the features.
