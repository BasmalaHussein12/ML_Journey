# Feature Extraction

Feature Extraction is the process of transforming raw data into meaningful and structured features that Machine Learning models can use effectively.

The main idea is to convert complex or raw information into a more useful representation while preserving important information.

## Why do we use Feature Extraction?

- Simplifies complex data.
- Reduces computational cost.
- Removes some noise or irrelevant information.
- Helps models learn useful patterns.
- Can reduce the number of features.
- Can improve model performance and generalization.

---

# Feature Selection vs Feature Extraction

## Feature Selection

Feature Selection chooses a subset of the original features.

Example:

Original features:

Age, Salary, Height, Weight, Experience

After Feature Selection:

Age, Salary, Experience

The selected features remain unchanged.

### Main idea:

Selection = Choose existing features.

---

## Feature Extraction

Feature Extraction transforms the original features into a new representation.

Example:

Original features:

X1, X2, X3, X4

After Feature Extraction:

New Feature 1, New Feature 2

The new features may represent information from multiple original features.

### Main idea:

Extraction = Create a new representation from existing information.

# Main Feature Extraction Techniques

## 1. Statistical Methods

Statistical methods summarize and describe patterns in data.

### Mean

The average value of a dataset.

### Median

The middle value after sorting the data.

### Standard Deviation

Measures how spread out the data is.

### Correlation

Measures the linear relationship between variables.

### Covariance

Measures how two variables change together.

---

# 2. Dimensionality Reduction

Dimensionality Reduction reduces the number of features while trying to preserve important information.

Example:

100 features

↓

Dimensionality Reduction

↓

10 features

Benefits:

- Reduces computation.
- Simplifies the dataset.
- Can help visualization.
- Reduces feature complexity.

### PCA — Principal Component Analysis

PCA transforms the original features into new components.

The components are designed to capture as much variance in the data as possible.

Main idea:

PCA → Preserve maximum variance.

### LDA — Linear Discriminant Analysis

LDA creates a new representation that focuses on separating different classes.

Main idea:

LDA → Maximize class separability.

### t-SNE

t-SNE reduces high-dimensional data into usually 2 or 3 dimensions.

It is especially useful for visualizing complex high-dimensional datasets.

Main idea:

t-SNE → Visualization of high-dimensional data.

---

# 3. Text Feature Extraction

Machine Learning models cannot directly understand raw text.

We convert text into numerical features.

## Bag of Words (BoW)

Bag of Words represents a document using word frequencies.

It counts how many times each word appears.

Important:

BoW ignores word order.

Main idea:

BoW → Word frequency.

---

## TF-IDF

TF-IDF represents the importance of words.

It considers:

TF → How frequently a word appears in a document.

IDF → How uncommon the word is across documents.

Main idea:

TF-IDF → Important words in a document that are relatively uncommon across the collection.

---

# 4. Signal Processing Methods

Signal processing can be used with:

- Audio
- Time-series data
- Sensor data

## Fourier Transform

Fourier Transform converts a signal from the time domain to the frequency domain.

Main idea:

Fourier Transform → Analyze frequency components.

## Wavelet Transform

Wavelet Transform analyzes signals that change over time.

It provides information about both time and frequency.

Main idea:

Wavelet Transform → Time + Frequency information.

---

# 5. Image Feature Extraction

Feature extraction can be used to represent important information from images.

## HOG — Histogram of Oriented Gradients

HOG extracts information about intensity gradients and edge directions in an image.

It can be used for:

- Object Detection
- Object Recognition

Main idea:

HOG → Edges + Directions.

---

## CNN Features

Convolutional Neural Networks can learn hierarchical features from images.

A simplified representation:

Image

↓

Edges

↓

Textures

↓

Shapes

↓

Complex patterns

↓

Object representation

CNNs can automatically learn useful image features through convolutional layers.
