# Data Cleaning

## What is Data Cleaning?

Real-world data is rarely perfect.

Before training a Machine Learning model, we need to inspect and clean the dataset because raw data usually contains errors, missing values, duplicates, inconsistent formats, or incorrect values.

**Data Cleaning** is the process of detecting and correcting these problems to make the dataset accurate, consistent, and suitable for analysis and Machine Learning.

---

## Why is Data Cleaning Important?

Poor-quality data leads to poor-quality models.

Data cleaning helps us:

* Improve data quality.
* Increase model accuracy.
* Reduce bias.
* Prevent misleading analysis.
* Make the dataset easier to understand.
* Prepare the data for EDA and Machine Learning.

> **Garbage In = Garbage Out (GIGO).**
>
> If the input data is poor, the model's predictions will also be poor.

---

# Common Data Quality Problems

### 1. Missing Values

Some information is unavailable.

Example

| Name  | Age |
| ----- | --- |
| Ahmed | 20  |
| Sara  | NaN |

Possible solutions:

* Remove the row.
* Replace with Mean.
* Replace with Median.
* Replace with Mode.
* Predict the missing value.

---

### 2. Duplicate Records

The same observation appears more than once.

Example

| Name  | Age |
| ----- | --- |
| Ahmed | 20  |
| Ahmed | 20  |

Duplicates may bias the model, so they should be removed.

---

### 3. Incorrect Data Types

Sometimes data is stored in the wrong format.

Example

Age:

```python
"20"
```

instead of

```python
20
```

A numerical value should not be stored as text.

---

### 4. Outliers

Outliers are observations that are significantly different from the rest of the data.

Example

```text
18
20
19
21
150
```

The value **150** is suspicious.

Outliers may represent:

* Data entry errors.
* Rare but valid observations.

Therefore, they should always be investigated before removing them.

---

### 5. Inconsistent Formats

The same information is represented in different ways.

Example

```text
2026-08-01
01/08/2026
August 1, 2026
```

Although they represent the same date, the dataset should use one consistent format.

---

### 6. Spelling and Typographical Errors

Example

```text
Alex
alex
ALEX
Alxe
```

To a computer, these are four different categories.

---

# Data Preprocessing

## What is Data Preprocessing?

Data Preprocessing is the process of preparing raw data before training a Machine Learning model.

Its main goal is to transform the dataset into a clean, structured, and machine-readable format so that learning algorithms can discover meaningful patterns effectively.

Data preprocessing is considered one of the most important stages in the Machine Learning pipeline because the quality of the input data directly affects the quality of the model.

---

## Why is Data Preprocessing Important?

A well-preprocessed dataset helps to:

* Improve model performance.
* Increase prediction accuracy.
* Reduce noise in the data.
* Speed up model training.
* Prevent misleading results.
* Make the data easier to analyze and visualize.
* Prepare the dataset for Machine Learning algorithms.

---

## Normalization (Min-Max Scaling)

Normalization rescales numerical values into a fixed range, usually between **0 and 1**.

Formula:

```text
X_scaled = (X - X_min) / (X_max - X_min)
```

Best suited for:

* k-Nearest Neighbors (KNN)
* Neural Networks

---

## Standardization (Z-score Scaling)

Standardization transforms data so that:

* Mean = 0
* Standard Deviation = 1

Formula:

```text
Z = (X - μ) / σ
```

where:

* **X** = Original value
* **μ** = Mean
* **σ** = Standard deviation

Best suited for:

* Linear Regression
* Logistic Regression
* Support Vector Machine (SVM)
* Principal Component Analysis (PCA)

---

## Understanding the Box Plot

A Box Plot is a visualization used to understand the distribution of numerical data and detect outliers.

It consists of five main parts:

- **Minimum:** The smallest non-outlier value.
- **Q1 (First Quartile):** 25% of the data lies below this value.
- **Median (Q2):** The middle value of the dataset (50%).
- **Q3 (Third Quartile):** 75% of the data lies below this value.
- **Maximum:** The largest non-outlier value.

The box represents the middle **50%** of the data (from Q1 to Q3).

The line inside the box represents the **Median**.

The whiskers extend to the minimum and maximum **non-outlier** values.

Any points plotted outside the whiskers are called **Outliers**, which are unusually high or low observations compared to the rest of the dataset.

### Why do we use a Box Plot?

- Detect outliers.
- Understand the spread of the data.
- Compare distributions between different features.
- Identify whether the data is skewed.

------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------

## Removing Outliers using IQR

Outliers are observations that lie far away from the majority of the data. They may represent data entry errors, measurement errors, or rare observations.

One of the most common techniques for detecting outliers is the **Interquartile Range (IQR)** method.

The IQR measures the spread of the middle **50%** of the data.

Formula:

```text
IQR = Q3 - Q1
```

Where:

- **Q1** = 25th percentile
- **Q3** = 75th percentile

To identify outliers, we calculate the acceptable range:

```text
Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR
```

Any value below the Lower Bound or above the Upper Bound is considered an **Outlier**.

### Steps

1. Calculate Q1 and Q3.
2. Compute the IQR.
3. Calculate the lower and upper bounds.
4. Remove observations outside the acceptable range.

This method is simple, effective, and widely used for preprocessing numerical features before training Machine Learning models.

````python
q1, q3 = np.percentile(df['Insulin'], [25,27])

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

clean_df = df[(df['Insulin'] >= lower) & (df['Insulin'] <= upper)]
````

- **-1** → Perfect Negative Correlation

In Machine Learning, correlation analysis helps us understand which features are most related to the target variable.

> **Important Note:**
>
> **Correlation does not imply causation.**
>
> A strong correlation between two variables does **not** necessarily mean that one variable causes the other. It simply indicates that they change together in a certain pattern.
>
> In other words:
>
> - **Correlation** answers the question: *"Are these two variables related?"*
> - **Causation** answers the question: *"Does one variable directly cause changes in the other?"*
>
> Therefore, correlation is useful for discovering relationships in the data, but it cannot be used as evidence of a cause-and-effect relationship.

# Step 8: Feature Scaling (Normalization)

Different features usually have different ranges.

Example:

| Feature | Value |
|---------|------:|
| Age | 25 |
| Glucose | 180 |
| Insulin | 850 |

Large differences in feature scales can negatively affect many Machine Learning algorithms.

Feature Scaling transforms numerical features into a common scale.

One of the most common techniques is **Min-Max Scaling (Normalization)**.

It rescales all values between **0 and 1**.

Formula:

```text
X_scaled = (X - X_min) / (X_max - X_min)
```

### Implementation

```python

scaler = MinMaxScaler()

X_normalized = scaler.fit_transform(X)

print(X_normalized[:5])
```

### Code Explanation

```python

```

Creates a scaler object from Scikit-learn.

```python
scaler.fit_transform(X)
```

Performs two operations:

- **fit()** → Learns the minimum and maximum value of every feature.
- **transform()** → Applies Min-Max Scaling and converts all values into the range [0, 1].

```python
print(X_normalized[:5])
```

Displays the first five normalized samples.

### Why do we normalize the data?

- Prevents features with large values from dominating the model.
- Improves training speed.
- Helps many Machine Learning algorithms perform better.
- Ensures all features contribute fairly during training.

> **Note:** Feature Scaling is applied only to the input features (**X**), not to the target variable (**y**) in classification problems.
