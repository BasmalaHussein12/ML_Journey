# Starting Machine Learning Journey: 6th of July 2026

### What exactly does "Machine Learning" mean?

Actually, after searching for it, I found that **machines** refer to devices such as computers or any system capable of performing specific tasks.

The word **learning** means that these machines improve their performance by learning from data instead of being explicitly programmed for every possible situation.

This idea led scientists to develop a branch of **Artificial Intelligence (AI)** called **Machine Learning (ML)**.

Instead of writing a rule for every problem, we provide the machine with data, allowing it to discover patterns, relationships, and knowledge on its own. It then uses what it has learned to make predictions or decisions when faced with new data.

This ability to learn from experience rather than relying solely on predefined rules is what makes Machine Learning one of the most powerful fields in Artificial Intelligence.


# Machine Learning Glossary

This section contains the most important Machine Learning concepts and definitions that I learned throughout my journey.

---

# Dataset

A **dataset** is a collection of data used to train, validate, and test a machine learning model.

---

# Sample (Observation)

A **sample** is a single row (or one example) in a dataset.

---

# Feature

A **feature** is an input variable used by the model to make predictions.

**Example**

| Age | Salary | Experience | Bought Product |
| :-: | :----: | :--------: | :------------: |
|  22 |  6000  |      1     |       Yes      |

Features:

* Age
* Salary
* Experience

---

# Label (Target)

A **label** (or **target**) is the correct output that the model is trying to predict.

In the previous example:

**Label:** Bought Product

---

# Model

A **model** is the mathematical representation that learns patterns and relationships from data in order to make predictions on unseen data.

---

# Prediction

A **prediction** is the output generated by a trained model when it receives new input data.

---

# Training

**Training** is the process of teaching a model using historical data so it can learn patterns and relationships.

---

# Validation

**Validation** is the process of evaluating a model during training to tune its hyperparameters and prevent overfitting.

---

# Testing

**Testing** is the final evaluation stage where the trained model is assessed using unseen data to measure its real-world performance.

---

# Machine Learning Pipeline

A **Machine Learning Pipeline** is an ordered sequence of steps where the output of one stage becomes the input to the next stage, from raw data to a deployed model.

Example:

```text
Raw Data
    ↓
Data Cleaning
    ↓
EDA
    ↓
Feature Engineering
    ↓
Model Training
    ↓
Evaluation
    ↓
Deployment
```

---

# Learning Type

A **learning type** describes **how the model learns from data** based on the information available during training.

The four main learning types are:

* Supervised Learning
* Unsupervised Learning
* Semi-Supervised Learning
* Reinforcement Learning

---

# Supervised Learning

Supervised Learning is a learning type in which the dataset contains both input data (features) and the correct outputs (labels). The model learns the relationship between them.

---

# Regression

Regression is a supervised learning task where the model predicts **continuous numerical values**.

Examples:

* House prices
* Temperature
* Sales

---

# Classification

Classification is a supervised learning task where the model predicts **categories or classes**.

Examples:

* Spam or Not Spam
* Cat or Dog
* Fraud or Normal

---

# Unsupervised Learning

Unsupervised Learning is a learning type where the dataset contains no labels. The model discovers hidden patterns and structures within the data.

---

# Clustering

Clustering is the process of grouping similar data points together without knowing the correct groups beforehand.

---

# Dimensionality Reduction

Dimensionality Reduction is the process of reducing the number of features while preserving as much useful information as possible.

---

# Association Rule Learning

Association Rule Learning discovers relationships and patterns between items in a dataset.

Example:
People who buy bread often buy butter.

---

# Anomaly Detection

Anomaly Detection identifies unusual or abnormal observations that differ significantly from the rest of the data.

Examples:

* Fraud Detection
* Network Intrusion
* Equipment Failure

---

# Semi-Supervised Learning

Semi-Supervised Learning uses a small amount of labeled data together with a large amount of unlabeled data during training.

---

# Reinforcement Learning

Reinforcement Learning is a learning type in which an agent learns by interacting with an environment and receiving rewards or penalties for its actions.

---

# Feature Engineering

Feature Engineering is the process of creating new features or transforming existing features to help the model learn better.

---

# Feature Selection

Feature Selection is the process of selecting only the most relevant features that contribute significantly to the model's performance.

---

# Domain Expertise

Domain Expertise is the use of specialized knowledge about a specific field to create, select, or improve features and make better machine learning decisions.

---

# Data Preprocessing

Data Preprocessing is the process of cleaning and preparing raw data before training a machine learning model.

It typically includes:

* Handling missing values
* Removing duplicates
* Encoding categorical variables
* Feature scaling
* Data cleaning

---

# Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is the process of analyzing and visualizing data to understand its structure, discover patterns, detect anomalies, and generate useful insights before building a model.

---

# Model Selection

Model Selection is the process of choosing the most suitable machine learning algorithm based on the problem, the nature of the data, and the desired outcomes.

---

# Model Evaluation

Model Evaluation is the process of measuring how well a trained model performs using evaluation metrics.

---

# Hyperparameter Tuning

Hyperparameter Tuning is the process of finding the best set of hyperparameters to improve a model's performance.

---

# Model Deployment

Model Deployment is the process of making a trained machine learning model available for real-world use.

---

# Monitoring and Maintenance

Monitoring and Maintenance is the continuous process of tracking a deployed model's performance and updating it whenever necessary to ensure reliable predictions over time.

---

# Underfitting

Underfitting occurs when a model is too simple to learn the underlying patterns in the training data, resulting in poor performance.

---

# Overfitting

Overfitting occurs when a model memorizes the training data instead of learning general patterns, causing poor performance on unseen data.

---

# Bias

Bias is the error introduced because a model is too simple or makes overly strong assumptions about the data.

---

# Variance

Variance is the error caused by a model being too sensitive to the training data, making it perform poorly on new data.

---

# Regularization

Regularization is a set of techniques used to reduce overfitting by preventing the model from becoming overly complex.

Examples:

* L1 Regularization (Lasso)
* L2 Regularization (Ridge)
* Dropout (Deep Learning)

---

# Cross-Validation

Cross-Validation is a model evaluation technique that repeatedly splits the dataset into different training and validation sets to estimate the model's performance more reliably.

---

# Hyperparameter

A **hyperparameter** is a configuration value set **before training** that controls how a machine learning algorithm learns.

Examples:

* Learning Rate
* Number of Neighbors (K)
* Tree Depth
* Number of Epochs

---

# Feature Scaling

Feature Scaling is the process of bringing numerical features to a similar scale so that no single feature dominates the learning process.

---

# Normalization

Normalization is a feature scaling technique that transforms values into a fixed range, usually between **0 and 1**.

---

# Standardization

Standardization is a feature scaling technique that transforms data to have:

* Mean = 0
* Standard Deviation = 1

---

# Data Leakage

Data Leakage occurs when information that should not be available during training accidentally becomes part of the training data, leading to unrealistically high model performance.


# Machine Learning Life Cycle

Machine Learning has a structured lifecycle, and actually, it's very logical.

It starts with defining a problem we want to solve, then collecting the data to help the machine learn.

When we get this data, it's not a bed of roses yet. We should clean it and preprocess it for the next stages.

Okay, we now have the machine and the data to learn. What else?

Machines learn by finding patterns and characteristics hidden in the data, and to achieve this, we use something called **Exploratory Data Analysis (EDA)**.

EDA is used to uncover insights and understand the dataset's structure.

---

## Step 1: Exploratory Data Analysis (EDA)

EDA helps us understand our dataset before training any model.

It allows us to:

* Discover hidden patterns.
* Understand relationships between variables.
* Detect missing values and outliers.
* Gain insights that guide the next stages.

---

## Step 2: Feature Engineering and Selection

After performing EDA, we move to a new stage called **Feature Engineering and Selection**.

Let's start with the word **Feature**.

A **feature** is an input variable.

### Example

| Age | Salary | Experience | Bought Product |
| :-: | :----: | :--------: | :------------: |
|  22 |  6000  |      1     |       Yes      |

So, the features are:

* Age
* Salary
* Experience

The label (target) is:

* Bought Product

### What is Feature Engineering?

Feature Engineering is the process of creating new features or transforming existing ones to help the model learn better.

This stage aims to create meaningful features and select only the relevant ones to improve model performance while reducing complexity.

---

## Step 3: Model Selection

After determining our features and making sure everything is ready, we move to the **Model Selection** process.

I see it as one of the most important parts of the whole lifecycle because our main goal is to solve the problem we defined.

Therefore, we need a model that aligns with:

* The problem itself.
* The nature of the data.
* The complexity of the problem.
* The desired outcomes.

---

## Step 4: Model Training

After selecting our model, we start training it to solve our problem.

This process involves exposing the model to historical data, allowing it to learn patterns, relationships, and dependencies within the dataset.

---

## Step 5: Model Evaluation and Tuning

While training the model, it might produce errors or make mistakes.

After all, it's just a machine that follows human instructions.

To overcome this, we move to the **Model Evaluation and Tuning** stage.

This stage provides insights into the model's strengths and weaknesses, helping us improve its performance before deployment.

---

## Step 6: Model Deployment

After solving these errors and getting satisfying results, we move to the **Model Deployment** process.

Now the model is ready to be deployed in real-world applications.

---

## Step 7: Monitoring and Maintenance

Did you think we were finished? 😂

Man, remember, you're dealing with a machine with no brain.

We should monitor and maintain our model after deployment to ensure it continues to perform well over time.

Data changes, user behavior changes, and the environment changes. Therefore, our model should be monitored regularly and updated whenever necessary.
