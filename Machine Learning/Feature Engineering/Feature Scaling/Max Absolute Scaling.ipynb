# Max Absolute Scaling

## 1. What is Max Absolute Scaling?

**Max Absolute Scaling** is a Feature Scaling technique that scales each feature by its maximum absolute value.

It transforms the data so that the values are generally within the range:

$$
[-1,1]
$$

### Formula

$$
x' = \frac{x}{\max(|x|)}
$$

Where:

- $x$ = original value
- $\max(|x|)$ = maximum absolute value of the feature
- $x'$ = scaled value


## 2. Why Do We Use Max Absolute Scaling?

Different features can have different scales.

Max Absolute Scaling puts features on a comparable scale without changing the sign of the values.

For example:

```text
Original:
[-100, -50, 0, 50, 100]

Scaled:
[-1, -0.5, 0, 0.5, 1]

````python
from sklearn.datasets import load_diabetes

data = load_diabetes(as_frame=True)

df = data.frame
df
````

## 5. Max Absolute Scaling vs Min-Max Scaling

### Min-Max Scaling

Uses:

$$
x' = \frac{x-x_{min}}{x_{max}-x_{min}}
$$

It shifts the values based on the minimum and maximum.

Default range:

```text
[0,1]


## 6. Important Notes

- `MaxAbsScaler` works on features/columns.
- Each feature is scaled independently.
- It divides by the maximum absolute value.
- It generally produces values between `-1` and `1`.
- It preserves the sign of the values.
- It preserves zero values.
- It does not center the data around zero.
- It is useful for sparse data.


## 7. Key Takeaways

- **Max Absolute Scaling** scales each feature using its maximum absolute value.
