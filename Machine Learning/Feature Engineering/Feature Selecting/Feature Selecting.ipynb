# Feature Selection

Feature selection is the process of choosing only the most useful features from the dataset for a machine learning model.

The main idea is simple:

> We have many features, but not all of them are useful for prediction.

Feature selection helps us remove irrelevant or redundant features.

### Why do we use Feature Selection?

- Remove irrelevant features
- Reduce noise
- Reduce model complexity
- Make training faster
- Improve interpretability
- Reduce the risk of overfitting
- Help the model focus on useful information

### Simple Example

Suppose we want to predict whether a person will buy a product:

| Age | Salary | Height | Hours_Studied | Bought |
|---|---:|---:|---:|---|
| 20 | 5000 | 170 | 5 | Yes |
| 25 | 7000 | 180 | 2 | No |
| 22 | 6000 | 165 | 6 | Yes |

Features:

- Age
- Salary
- Height
- Hours_Studied

Target:

- Bought

If `Height` has almost no relationship with `Bought`, we may remove it.

So instead of:

Age + Salary + Height + Hours_Studied

we may use:

Age + Salary + Hours_Studied

This is Feature Selection.
