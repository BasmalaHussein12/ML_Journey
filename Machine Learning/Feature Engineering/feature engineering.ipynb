# Feature Engineering

Feature Engineering is the process of selecting, creating, transforming, or modifying
features so that Machine Learning models can learn patterns more effectively.

The main goal:
Turn raw data → useful features → better ML model

## Why do we use Feature Engineering?

Feature Engineering can:

- Improve model accuracy
- Reduce overfitting
- Make models easier to understand
- Make training and prediction more efficient

Think of it as:

Raw / messy data
        ↓
Feature Engineering
        ↓
Useful features
        ↓
Machine Learning model

# Main Processes of Feature Engineering

Feature Engineering can involve:

1. Feature Creation
2. Feature Transformation
3. Feature Extraction
4. Feature Selection
5. Feature Scaling

These techniques are related, but they do different jobs.

## 1. Feature Creation

Feature Creation means creating a NEW feature from existing information.

Example:

If we have:

height
weight

We can create:

BMI = weight / height²

The original features still exist, but we created an additional useful feature.

```python
import pandas as pd

df = pd.DataFrame({
    "height": [1.60, 1.75, 1.80],
    "weight": [55, 70, 85]
})

df["BMI"] = df["weight"] / (df["height"] ** 2)

df
```

## 2. Feature Transformation

Feature Transformation changes the representation of an existing feature.

Examples:

- Scaling
- Normalization
- Log transformation
- Encoding categorical variables

The goal is to make the data easier for the ML model to work with.

For example:

Age:       20, 30, 40, 50

can be transformed/scaled into:

0.0, 0.33, 0.67, 1.0

The information is still representing age, but its numerical representation changed.

## 3. Feature Extraction

Feature Extraction transforms existing information into a NEW representation
that contains useful information.

Examples:

- PCA
- Bag of Words
- TF-IDF
- HOG
- CNN features

Example:

100 original features
        ↓
      PCA
        ↓
10 new features

The new features are usually different representations of the original data.

Feature Selection
→ choose existing features

Feature Extraction
→ create a new representation from existing features

## 4. Feature Selection

Feature Selection means choosing the most useful features
and removing irrelevant or redundant ones.

Example:

Original:

Age
Salary
Height
Weight
Favorite_Color
Random_ID

After selection:

Age
Salary
Weight

We KEEP useful existing features.
We REMOVE unnecessary ones.

## 5. Feature Scaling

Feature Scaling puts numerical features on comparable scales.

Example:

Age:      20 → 60
Salary:   5000 → 50000

Salary has much larger numerical values.

Some ML algorithms can be affected by this difference.

Scaling can transform them into comparable ranges.

# One-Hot Encoding

Categorical features contain categories instead of numerical values.

Example:

Color

-----
Red
Blue
Green

Machine Learning models generally need numerical input.

One-Hot Encoding converts categories into binary columns.

```python
import pandas as pd 

df = pd.DataFrame({
    "color": ["red", "blue", "green", "blue"]
})

df_encoded = pd.get_dummies(
     df,
     columns=["color"],
     prefix="color"   
)

df_encoded
```

```python
df_encoded = pd.get_dummies(
     df,
     columns=["color"],
     prefix="color",
     dtype=int   
)

df_encoded
```

red → [0, 0, 1]

blue → [1, 0, 0]

green → [0, 1, 0]

One-Hot Encoding = categorical values → binary columns

# Binning

Binning converts continuous numerical values into groups/categories.

Example:

Age:

18
23
34
45
67

can become:

0-20
21-40
21-40
41-60
61+

Why?

Sometimes groups are more useful for analysis or modeling than exact values.

```python
import pandas as pd 

df = pd.DataFrame({
    "age":[23, 45, 18, 34, 67, 50, 21] 
})

bins = [0, 20, 40, 60, 100]
labels = ["0-20", "21-40", "41-60", "61+"]

df["age_group"] = pd.cut(
      df["age"],
      bins=bins,
      labels=labels,
      right= False
)

df
```

Binning = continuous values → groups

# Text Data Preprocessing

Machine Learning cannot directly work with raw text in the same way it works
with numerical features.

Text preprocessing can involve:

- Removing stop words
- Stemming
- Vectorization

Example:

"This is a sample sentence"

        ↓ preprocessing

"sampl sentenc"

        ↓ vectorization

Numerical representation

## Vectorization

Vectorization converts text into numerical features.

One example is:

Bag of Words

It represents text using word frequencies.

Another important technique:

TF-IDF

It gives more importance to words that are useful/distinctive
and less importance to very common words.

# Feature Splitting

Feature Splitting means taking one feature and dividing it
into multiple useful features.

Example:

Full_Address

"123 Elm St, Springfield, 12345"

can become:

Street       → 123 Elm St
City         → Springfield
Zipcode      → 12345

Why?

Because different parts may contain different useful information.

```python
import pandas as pd

df = pd.DataFrame({
    "Full_Address": [
        "123 Elm St, Springfield, 12345",
        "456 Oak Rd, Shelbyville, 67890"
    ]
})

df[["Street", "City", "Zipcode"]] = df["Full_Address"].str.extract(
    r'([0-9]+\s[\w\s]+),\s([\w\s]+),\s(\d+)'
)

df
```

Feature Splitting = one feature → multiple useful features

# The Most Important Differences

Feature Selection

Feature Extraction

Feature Scaling
→ Put numerical features on comparable scales.

One-Hot Encoding
→ Convert categorical values into numerical binary columns.

Binning
→ Convert continuous values into groups.

Feature Splitting
→ Divide one feature into multiple features.
