# Ticks & Styles

In this notebook, I learn how to customize axis ticks and apply different styles to make plots more readable and visually appealing.

## Topics
- Ticks
- Tick Labels
- Tick Rotation
- Styles
- rcParams

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [120, 150, 140, 180, 200]

plt.plot(months, sales, marker ="o")
plt.show()
```

# What are Ticks?

Ticks are the labels shown on the x-axis and y-axis

The numbers (120, 130, 140...) and month names (Jan, Feb...) are ticks

```python
plt.plot(months, sales)

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.show()
```

```python
plt.plot(months, sales)

plt.xticks(rotation=45)
plt.show()
```

```python
numbers = [1, 2, 3, 4, 5]
values = [10, 20, 30, 40, 50]

plt.plot(numbers, values)
plt.xticks([1,3,5])

plt.show()
```

Only 1, 3, and 5 will appear on the x-axis.

```python
plt.style.use("ggplot")

plt.plot(months, sales)

plt.show()
```

# Use rcParams

rcParams changes the default settings for all plots.

```python
plt.rcParams["figure.figsize"] = (8,5)

plt.rcParams["font.size"] = 12
```

Now every new plot will use these defaults.

```python
plt.style.use("ggplot")

plt.figure(figsize=(8,5))

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```
