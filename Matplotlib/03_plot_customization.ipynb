# Plot Customization

In this notebook, I learn how to customize Matplotlib plots to make them more readable and visually appealing.

## Topics
- Titles
- Axis Labels
- Grid
- Axis Limits
- Figure Size
- Fonts

```python
import matplotlib.pyplot as plt 

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 130, 180, 200]

plt.plot(months, sales)
plt.show()
```

```python
plt.plot(months, sales)

plt.title("Monthly Sales")

plt.show()
```

```python
plt.plot(months, sales)

print(type(plt.xlabel))
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
plt.plot(months, sales)

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize =(8,5))
# Width = 8 inches and Height = 5 inches

plt.plot(months, sales)
plt.show()
```

```python
plt.plot(months, sales)
plt.xlim("Jan", "Feb")
plt.ylim(100, 220)

plt.show()
# zoom in a specific range
```

```python
plt.plot(months, sales)

plt.title("Monthly sales", fontsize = 18)
plt.xlabel("Month", fontsize = 14)
plt.ylabel("Sales", fontsize = 14)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    months,
    sales,
    color="blue",
    marker="o",
    linestyle="-",
    label="Sales"
)

plt.title("Monthly Sales Report", fontsize=18)

plt.xlabel("Month", fontsize=14)

plt.ylabel("Sales", fontsize=14)

plt.grid(True)

plt.legend()

plt.show()
```

## Summary

In this notebook, I learned how to:

- Add titles
- Label the x-axis and y-axis
- Add a grid
- Change the figure size
- Set axis limits
- Adjust font sizes
- Combine multiple customizations into one professional plot
