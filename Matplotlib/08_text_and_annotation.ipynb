# Text & Annotation

In this notebook, I learn how to add text and annotations to Matplotlib plots to highlight important information.

## Topics
- plt.text()
- plt.annotate()
- Arrows
- Highlighting important points

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [120, 150, 140, 180, 200]

plt.plot(months, sales, marker="o")

plt.show()
```

```python
plt.plot(months, sales, marker="o")

plt.text("Mar", 140, "March")

plt.show()
```

# What does plt.text() do?

Syntax:

- plt.text(x, y, "Your Text")

It places text at a specific coordinate.

```python
plt.plot(months, sales, marker="o")

for i in range(len(months)):
    plt.text(months[i], sales[i],sales[i])

plt.show()    
```

Each point will display its value.

```python
plt.plot(months, sales, marker="o")

plt.annotate(
    "Highest Sale",
    xy=("May", 200),
    xytext=("May", 220),
    arrowprops=dict(arrowstyle="->")
)

plt.show()
```

Explanation
- xy → The point you want to highlight.

- xytext → Where the annotation text will appear.

- arrowprops → Draws an arrow.

```python
plt.plot(months, sales, marker="o")

plt.annotate(
    "Highest Sale",
    xy=("May",200),
    xytext=("Feb",210),
    arrowprops=dict(arrowstyle="->", color = "red")
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(months, sales, marker="o", color="blue")

plt.title("Monthly Sales")

plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.annotate(
    "Best Month",
    xy=("May", 200),
    xytext=("Mar", 210),
    arrowprops=dict(
        arrowstyle="->",
        color="red"
    ),
     color="green"
)

plt.show()
```
