# Legends & Colors

In this notebook, I learn how to use legends and colors to create clear and professional visualizations.

## Topics
- Legends
- Colors
- Color Maps (Colormaps)
- Colorbars
- Choosing the Right Colors

```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [120, 150, 140, 180, 200]
profit = [30, 40, 35, 50, 60]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.legend()
plt.show()
```

# What is a Legend?

A legend explains what each color or line represents.

The label parameter names each line, and plt.legend() displays those names.

```python
plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.legend(loc="upper right")
plt.show()
```

# Common Locations 

- loc="upper left"

- loc="upper right"

- loc="lower left"

- loc="lower right"

- loc="center"

- loc="best"

"best" lets Matplotlib automatically choose the best position.

```python
plt.plot(months, sales, color = "green" ,label="Sales")
plt.plot(months, profit, color ="skyblue",label="Profit")

plt.legend(loc="upper right")
plt.show()
```

# Some common colors:

- "red"

- "blue"

- "green"

- "black"

- "purple"

- "orange"

- "pink"

- "gray"

```python
plt.plot(months, sales, color="#1f77b4")
plt.show()
```

# Instead of color names, you can use HEX codes

Example:

#FF0000 → Red

#00FF00 → Green

#0000FF → Blue

```python
x = np.arange(10)
y = np.random.randint(1,20,10)

plt.scatter(
    x,
    y,
    c=y,
    cmap= "viridis",
    s=100
)

plt.show()
```

```python
scatter= plt.scatter(
    x,
    y,
    c=y,
    cmap= "viridis",
    s=100
)

plt.colorbar(scatter)

plt.show()
```

A colorbar explains what each color represents.
