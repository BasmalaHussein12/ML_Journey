# Line Plots

In this notebook, I learn how to create and customize line plots using Matplotlib.

## Topics
- plt.plot()
- Multiple Lines
- Colors
- Line Styles
- Markers
- Figure and Axes

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 18]

plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, color ="red")
# to change the line's color
```

```python
plt.plot(x, y, linestyle = "--")
# to change the line style
# "-"       Solid
# "--"      Dashed
# ":"       Dotted
# "-."      Dash-dot
```

```python
plt.plot(x, y, marker = "o")
plt.show()
# Add markers
# "s"   Square
# "^"   Triangle
# "*"   Star
# "x"   Cross
```

```python
plt.plot(x, y, marker = "*")
plt.show()
```

```python
plt.plot(
    x,
    y,
    color = "green",
    linestyle = "--",
    marker = "o"
)
plt.show()
```

```python
months = [1, 2, 3, 4, 5]

sales_2024 = [100, 120, 140, 160, 180]
sales_2025 = [110, 135, 150, 170, 200]

plt.plot(months, sales_2024)
plt.plot(months, sales_2025)

plt.show()
```

But now how to know each line is 2024 or 2025?

```python
plt.plot(months, sales_2024, label="2024")
plt.plot(months, sales_2025, label="2025")

plt.legend()
# display the labels for each line

plt.show()
```

# Remember
Figure = the whole image.
Axes = the plotting area inside that image.
Plot = the graph drawn on the axes.
