# Error Bars

Sometimes your measurements have uncertainty

```python
x = [1, 2, 3, 4, 5]

y = [10, 15, 13, 18, 20]

error = [1, 2, 1, 2, 1]
```

Create an error bar plot:

```python
plt.errorbar(
    x,
    y,
    yerr=error,
    fmt="o-",
    capsize=5
)

plt.title("Measurements with Error Bars")

plt.show()
```

# Explanation

yerr → Error values

fmt="o-" → Circle markers connected by lines

capsize → Width of the small horizontal caps

# When do we use Error Bars?

- Scientific experiments

- Sensor measurements

- Machine learning model accuracy with uncertainty

- Laboratory data
