```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 130, 180, 200]
profit = [20, 30,25,40, 45]

plt.subplot(1,2,1)
# subplot(rows, columns, position)
# so position = 1 = first plot

plt.plot(months, sales)
plt.title("Sales")

plt.subplot(1,2,2)

plt.plot(months, profit)
plt.title("Profit")
```

```python
plt.figure(figsize=(6,8))

plt.subplot(2,1,1)
plt.plot(months, sales)
plt.title("Sales")

plt.subplot(2,1,2)
plt.plot(months, profit)
plt.title("Profit")
```

```python
fig, ax = plt.subplots(1,2,figsize=(10,4))
```

This creates:
- 1 row 
- 2 columns

```python
fig, ax = plt.subplots(1,2,figsize=(10,4))

ax[0].plot(months, sales)
ax[0].set_title("Sales")

ax[1].plot(months, profit)
ax[1].set_title("Profit")

plt.show()
```

# Why use Subplot()?

It's:

- Cleaner
- Easier to orginze 
- Preferred for professional projects 
- Better when working with many plots 

```python
fig, ax = plt.subplots(1,2,figsize=(10,4))

ax[0].plot(months, sales)
ax[0].set_title("Line Plot")

ax[1].scatter(months, sales)
ax[1].set_title("Scatter Plot")

plt.show()
```
