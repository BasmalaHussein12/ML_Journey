# Histograms

In this notebook, I learn how to visualize the distribution of numerical data using histograms.

## Topics
- plt.hist()
- Bins
- Density
- Data Distribution
- Histogram vs Scatter Plot

```python
import matplotlib.pyplot as plt

grades = [60, 65, 70, 72, 75, 78, 80, 82, 85, 87, 90, 92, 95]

plt.hist(grades)
plt.show()
# A histogram groups the values into ranges and shows how many values fall into each range.
```

```python
plt.hist(grades, bins = 5)
plt.show()
# Bins are intervals that group numerical values in a histogram
# The height of each bar shows how many values are inside that interval
```

```python
plt.hist(grades, bins = 3, color ="skyblue")
plt.show()
```

```python
plt.hist(
    grades,
    bins = 5,
    color = "green",
    edgecolor = "black"
)
plt.show()
```

```python
plt.hist(grades, bins = 4, density = True)
plt.show()
```

density=False (default): shows counts.

density=True: shows probabilities (normalized distribution).

```python
plt.figure(figsize = (8,5))

plt.hist(
    grades,
    bins=5,
    color="red",
    edgecolor="black"
)

plt.title("Distribution of Student Grades")
plt.xlabel("Grades")
plt.ylabel("Frequency")

plt.grid(True)

plt.show()
```
