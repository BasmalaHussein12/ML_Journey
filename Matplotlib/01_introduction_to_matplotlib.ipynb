# Why Do We Use Matplotlib?
We use Matplotlib to:

- Visualize data
- Discover patterns and trends
- Compare values
- Detect outliers
- Present results professionally

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [120, 180, 150, 220]

plt.plot(months, sales)
plt.show()
```

Awesome! 
it represents these lists in a graph

# Matplotlib vs Seaborn

Matplotlib                                                Seaborn

- Low-level library                                       - High-level library

- Better default style                                    - More customizable
                                       

- Supports many plot types                                - Built for statistical visualization

- Foundation of Seaborn                                   - Built on top of Matplotlib


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [2, 4 ,6, 8]

plt.plot(x, y)
plt.show()
```

<!-- What happened? -->

x → values on the x-axis
y → values on the y-axis
plot() → creates the graph
show() → displays it

<!-- pyplot Interface -->
The easiest way to create plots.

# plt.plot(x, y)
# plt.show()

Best for:
- Learning
- Quick visualizations
- Small projects

<!-- Object-Oriented Interface -->

# fig, ax = plt.subplots()
# ax.plot(x, y)
# plt.show()

<!-- What are fig and ax? -->
Figure → the entire canvas (the whole image)
Axes → the area where the graph is drawn

<!-- Saving Figures -->
