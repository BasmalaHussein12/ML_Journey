# Scatter Plots

In this notebook, I learn how to create scatter plots using Matplotlib and understand when to use them for exploring relationships between variables.

## Topics
- plt.scatter()
- Scatter vs Line Plot
- Marker Size
- Colors
- Transparency (alpha)
- Data Analysis Applications

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [50, 55, 60, 68, 75, 82, 90, 95]

plt.scatter(hours, scores)
plt.show()
```

<!-- Each point represents one observation. -->

Example:
Student 1 → (1, 50)
Student 2 → (2, 55)
Student 3 → (3, 60)

```python
plt.scatter(hours, scores)

plt.title("Study Hours VS Exam Scores")

plt.xlabel("Study hours")
plt.ylabel("Exam scores")

plt.show()
```

```python
plt.scatter(hours, scores, color = "red")

plt.show()
```

```python
plt.scatter(hours, scores, s=120)
plt.show()
# s controls the marker size.
```

```python
plt.scatter(hours, scores, alpha=0.5)
plt.show()

# Transparency is useful when many points overlap.
# alpha = 1 → Fully visible
# alpha = 0.5 → Semi-transparent
# alpha = 0 → Invisible
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    hours,
    scores,
    color="blue",
    s=120,
    alpha=0.7
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)

plt.show()
```

```python
plt.plot(hours, scores)
plt.show()
# line plot
```

```python
plt.scatter(hours, scores)
plt.show()
# scatter plot
```

# Why Scatter Plots Matter in ML?

Scatter plots help answer questions like:

- Do students who study more get higher grades?
- Is there a relationship between height and weight?
- Does experience affect salary?

They are often the first visualization used during Exploratory Data Analysis (EDA).
