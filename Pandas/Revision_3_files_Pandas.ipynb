```python
What is Pandas?

Pandas is a Py library used to:
- Read datasets
- Organize data 
- Clean data 
- Analysis data 
- Prepare data for ML 

It mainly provides 2 data structure: 

__Series__
A one-dimensional data structure(like a single column)
```

```python
import pandas as pd 

grades = pd.Series([50,69,90,34,85])
print(grades)
```

```python
__DataFrame__
A two-dimensional table with rows and columns 
```

```python
import pandas as pd

students = {
      "Names": ["Basmala", "Tasneem", "Mariam"],
      "Grades": [95, 55,87],
      "Ages": [19, 14, 9]
}

df = pd.DataFrame(students)
print(df)
```
