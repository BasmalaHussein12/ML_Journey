```python
# Selecting & Filtering Data

In this notebook, I learn how to select rows and columns, filter data using conditions, and sort datasets in Pandas.

## Topics
- Selecting columns
- Selecting rows
- loc
- iloc
- Boolean Indexing
- Sorting
- value_counts()
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df 
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.loc[0]
# loc selects rows by their labels (index)
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.loc[[0,2,3]]
# to display multiple rows by loc
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.iloc[0]
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.iloc[0:3, 1:3]
# 0:3 --> from row 0 to 2
# 1:3 --> from column 1 to 2
```

```python
loc → selects by label (index name)
iloc → selects by position (row/column number)
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df[df["Grade"]> 90]
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df[df["Age"]<21]
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df[df["City"]== "Cairo"]
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df[(df["Grade"]> 85) & (df["Age"]< 21)]
# multiple conditions with & (and) 
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df[(df["City"] == "Cairo") | (df["Grade"]> 90)]
# multiple conditions with | (OR) 
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.sort_values("Grade")
# to display the grades in an ascending order
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.sort_values("Grade", ascending = False)
# to display the grades in a descending order
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.sort_values("Age")
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df["City"].value_counts()
# to count values
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df["Grade"].value_counts()
# to count values
```
