```python
# Grouping & Aggregation

In this notebook, I learn how to group data and calculate statistics for each group.

## Topics
- groupby()
- count()
- sum()
- mean()
- max()
- min()
- agg()
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df
```

```python
# What is groupby()?
groupby("column's name") divides the dataset into groups based on a column, then lets you perform calculations on each group.
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")
# This will create groups but doesn't display results
```

```python
If I want to Count Students in Each City
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City").count()
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")["Name"].count()
# count name only 
```

```python
If I want Average Grade in Each City
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")["Grade"].mean()
```

```python
If I want Highest Grade in Each City
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")["Grade"].max()
```

```python
If I want Lowest Grade in Each City
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")["Grade"].min()
```

```python
If I want Total Grade in Each City
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")["Grade"].sum()
```

```python
If I want Multiple Statistics Using agg()
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("City")["Grade"].agg(["count", "mean", "max", "min", "sum"])
```

```python
If I want Group by Another Column
```

```python
import pandas as pd

df = pd.read_csv("students.csv")
df.groupby("Name")["Age"].mean()
```

```python
# So groupby() means:
Split the data into groups → Apply a calculation to each group → Return the result.
```
