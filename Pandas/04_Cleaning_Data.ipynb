```python
# Cleaning Data

In this notebook, I learn how to clean datasets by handling missing values, removing duplicates, renaming columns, and changing data types.

## Topics
- Missing Values
- isna()
- dropna()
- fillna()
- Duplicates
- drop_duplicates()
- rename()
- astype()
```

```python
1) If I want to check if there is any missing values 
---> df.isna()
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.isna()
```

```python
This returns a DataFrame of True and False
True → Missing value
False → Value exists
```

```python
2) To count the missing values 
---> df.isna().sum()
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.isna().sum()
```

```python
3) If I want to remove the rows that contain missing values 
---> df.dropna()
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.dropna()
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df = df.isna()
# to save the result
```

```python
4) To fill the missing values instead of removing 
---> df["column's name"].fillna(df["column's name"].mean())
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df = df["Grade"].fillna(df["Grades"].mean())
# .mean() ---> to will missing grades with mean 
# There are no missing values, so it raises an error
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df["City"] = df["City"].fillna("Unknowns")
# fill the missing values in city column by unknown
```

```python
5) Check for duplicates 
---> df.duplicated()
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.duplicated()
```

```python
Returns:
True → Duplicate row
False → Unique row
```

```python
6) To count duplicates:
---> df.duplicated().sum()
```

```python
7) To remove duplicates:
---> df = df.drop_duplicates()
```

```python
8) To rename columns:
---> df = df.rename(columns={
    "old column's name": "New name"
})
Before, I have:   Name  Age  Grade
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df = df.rename(columns={
    "Grade": "Final Grade"
})
df
```

```python
9) To change data type 
---> a) check the data type first by df.dtypes
b) convert:
df["Column's name"] = df["the same column's name"].astype(data type I want)
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df.dtypes
```

```python
import pandas as pd 

df = pd.read_csv("students.CSV")
df["Age"] = df["Age"].astype(float)
df.dtypes
```

```python
# Tips:

- isna() checks for missing values

- dropna() removes missing values

- fillna() replaces missing values

- duplicated() finds duplicates

- drop_duplicates() removes duplicates

- rename() changes column names

- astype() changes the data type of a column
```
