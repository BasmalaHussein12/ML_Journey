# Merging & Exporting Data

In this notebook, I learn how to combine multiple datasets and export the final results.

## Topics
- concat()
- merge()
- join()
- to_csv()
- to_excel()

```python
import pandas as pd 

student = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Basmala", "Tasneem", "Mariam"],
})

grade = pd.DataFrame({
    "ID": [1, 2, 3],
    "Grade": [90,75,88]
})

student
```

```python
grade
```

<!-- Concatenate DataFrames -->
Concatenate means placing DataFrames on top of each other or side by side.

 # pd.concat()
 <!-- Vertical -->

```python
pd.concat([student, grade])
```

```python
pd.concat([student, grade], axis=1)
# Horizontal
```

<!-- Merge DataFrames -->
Merge combines DataFrames using a common column.

```python
merged = pd.merge(student, grade, on="ID")

merged
```

<!-- Join DataFrames -->
Set the index first:

```python
students = student.set_index("ID")
grades = grade.set_index("ID")

students.join(grades)
```

```python
merged.to_csv("students_report.csv", index=False)
# This creates a CSV file in your current folder
```

```python
merged.to_excel("students_report.xlsx", index=False)
# This creates an Excel file.
# If you get an error saying that an Excel writer engine is missing, you can install the required package later (such as openpyxl).
```

## Summary

In this notebook, I learned how to:

- Combine DataFrames using `concat()`
- Merge DataFrames using `merge()`
- Join DataFrames using `join()`
- Export data to CSV
- Export data to Excel
