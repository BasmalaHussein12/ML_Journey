```python
import sklearn
sklearn.__version__
```

```python
import numpy as np

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

y = np.array([
    2,
    4,
    6,
    8,
    10
])
```

# Our relationship is basically:
- ## y = 2x

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X,y)
```

The model learns the relationship between X and y

```python
X_new = np.array([
    [6],
    [7],
    [8]
])

predictions = model.predict(X_new)
predictions
```

OMG! I am just trained a ML model

# Summary

In this notebook, I learned:

- Scikit-learn is a Python machine learning library.
- It provides tools for preprocessing, modeling, evaluation, and model selection.
- Scikit-learn uses a consistent API across many machine learning algorithms.
- Estimators use `fit()` to learn from data.
- Predictors use `predict()` to make predictions.
- Transformers use `transform()` to transform data.
- `fit_transform()` combines `fit()` and `transform()`.
- Training data is used to learn parameters.
- Learned transformations can then be applied to new data.
- Scikit-learn makes it easier to build machine learning workflows without implementing algorithms from scratch.

## Key workflow:

Data → Fit → Predict → Evaluate
