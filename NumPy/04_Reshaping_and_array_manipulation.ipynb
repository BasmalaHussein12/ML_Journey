```python
# flatten() and Ravel()
Convert a 2D array into a 1D copy
```

```python
import numpy as np

arr = np.array([
    [1,2,3],
    [4,5,6]
])

arr.flatten()
```

```python
import numpy as np

arr = np.array([
    [1,2,3],
    [4,5,6]
])

arr.ravel()
```

```python
What's the difference then??

flatten() returns a copy
ravel() usually returns a view, so changes may affect the original array
```

```python
# transpose() or .T
It reverses rows and columns 

(ya3ny el rows become columns w el columns become rows)
```

```python
import numpy as np

arr = np.array([
    [1,2,3],
    [4,5,6]
])

arr.transpose()
```

```python
import numpy as np

arr=np.array([[1, 2],
       [3, 4],
       [5, 6]])

arr.T
```

```python
If I have 2 arrays, and I wanna concatenate
I will use comma (,)
```

```python
import numpy as np

arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

np.concatenate((arr1,arr2)) 
```

```python
# Creating arrays from existing arrays:
---stack---
1) np.vstack (v for vertical)
If we have 2 arrays, it stacks them on top of each other  

2) np.hstack (h for horizontal)
It puts them beside each other 
```

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

np.vstack((a, b))
```

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

np.hstack((a, b))
```
