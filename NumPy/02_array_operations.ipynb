```python
# In this notebook, I learn how to perform mathematical operations on NumPy arrays.

Topics:
- Arithmetic operations
- Universal functions (ufuncs)
- Aggregation functions
- Comparison operations
```

```python
# Arithmetic Operations
import numpy as np
# Create arrays:
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
# Addition:
a + b
```

```python
# Operations with Scalars
A scalar is a single number

a + 10

NumPy adds 10 to every element:
[11 12 13 14]
This is called broadcasting
```

```python
# Universal Functions (ufuncs)
NumPy provides built-in mathematical functions

- Square root:
np.sqrt(a)

- Power:
np.power(a, 2)

- Exponential:
np.exp(a)
```

```python
# Aggregation Functions
Working with all elements

Sum:
np.sum(a)

Mean:
np.mean(a)

Maximum:
np.max(a)

Minimum:
np.min(a)

Standard deviation:
np.std(a)
```

```python
# Comparison Operations
a = np.array([1, 2, 3, 4])

a > 2
```
