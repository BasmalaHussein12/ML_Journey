```python
# 2D Arrays
2-dimensional arrays represent tables or matrices 
-->* They are very important in ML because datasets are usually stored as matrices 
```

```python
import numpy as np
arr_2d = np.array([
    [1,2,3],
    [4,5,6]
])

arr_2d
```

```python
# Understanding shape

import numpy as np
arr_2d = np.array([
    [1,2,3],
    [4,5,6]
])

arr_2d.shape
```

```python
# explanation
Shape tells us the dimensions of the array 
---> (2,3) means 2 rows and 3 columns
```

```python
# Understanding size

import numpy as np
arr_2d = np.array([
    [1,2,3],
    [4,5,6]
])

arr_2d.size
```

```python
# explanation 
Size returns the total number of elements in the array
```

```python
# dtype 
Like in normal Python 
tell us the type of data stored inside the array 

Examples:
- int 
- float 
- bool 
```

```python
## Mini Exercise 

Create a NumPy array representing the grades of 5 students

find:
- Array type
- shape 
- Number of elements 
- Data Type 
```

```python
import numpy as np 

grades = np.array([90,85,70,65,50])

print(type(grades))
print(grades.shape)
print(grades.size)
print(grades.dtype)
```

```python
### I noticed an important note 
There is a difference between type() and .dtype
- type() ----> returns array type 
- dtype ----> returns data type
```
